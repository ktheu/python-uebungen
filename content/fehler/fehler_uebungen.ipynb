#### Aufgabe A13
```
Der folgende Code soll das größte Element einer Liste finden, liefert
bei dieser Liste aber ein falsches Ergebnis. Finde den Fehler und
korrigiere ihn.
```
**Erwartete Ausgabe:**
```

```

````python
def find_maximum(a):
    best = 0
    for x in a:
        if x > best:
            best = x
    return best

numbers = [-8, -2, -15, -6]
result = find_maximum(numbers)
print(result)
````

#### Aufgabe A21
```
Die Funktion bubble_sort sortiert eine Liste von Zahlen aufsteigend.
Ändere sie so ab, dass sie eine Liste von Wörtern nach ihrer Länge
sortiert (das kürzeste Wort zuerst).
```
**Erwartete Ausgabe:**
```
['Ei', 'Brot', 'Apfel', 'Banane']
```

````python
def bubble_sort(a):
    getauscht = True
    while getauscht:
        getauscht = False
        for i in range(len(a) - 1):
            if a[i] > a[i + 1]:
                a[i], a[i + 1] = a[i + 1], a[i]
                getauscht = True

woerter = ['Banane', 'Ei', 'Apfel', 'Brot']
bubble_sort(woerter)
print(woerter)
````
