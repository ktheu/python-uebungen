```python
import sqlite3 as sq

def tabellen_anlegen():
    connection = sq.connect("sportverein.db")
    cursor = connection.cursor()

    cursor.execute('DROP TABLE IF EXISTS mitglied')  
    cursor.execute('DROP TABLE IF EXISTS bank;')
     
    cursor.execute('DROP TABLE IF EXISTS uebungsgruppe')
    cursor.execute('DROP TABLE IF EXISTS trainingsort')
    cursor.execute('DROP TABLE IF EXISTS trainingseinheit')
    cursor.execute('DROP TABLE IF EXISTS hausmeister')
    cursor.execute('DROP TABLE IF EXISTS teilnehmer')

    # ---- Tabelle bank --------------
    sql = """
    CREATE TABLE bank (blz INTEGER PRIMARY KEY, bezeichnung TEXT)
    """
    cursor.execute(sql)

    # ---- Tabelle mitglied -------
    sql = """
    CREATE TABLE mitglied (mid INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT, vorname TEXT, telefon TEXT,
    konto TEXT, blz INTEGER,
    beitrag REAL,
    FOREIGN KEY(blz) REFERENCES bank(blz))
    """ 
    cursor.execute(sql)

    # ---- Tabelle uebungsgruppe --------------
    sql = """
    CREATE TABLE uebungsgruppe (uid INTEGER PRIMARY KEY AUTOINCREMENT,
    sportart TEXT, mid INTEGER,
    FOREIGN KEY(mid) REFERENCES mitglied(mid))
    """
    cursor.execute(sql)

    # ---- Tabelle teilnehmer  = nimmt teil--------------
    sql = """
    CREATE TABLE teilnehmer (mid INTEGER, uid INTEGER,
    PRIMARY KEY(mid, uid),
    FOREIGN KEY(mid) REFERENCES mitglied(mid),
    FOREIGN KEY(uid) REFERENCES uebungsgruppe(uid))
    """
    cursor.execute(sql)

    # ---- Tabelle trainingsort  --------------
    sql = """
    CREATE TABLE trainingsort (ort TEXT PRIMARY KEY, hid INTEGER,
    FOREIGN KEY(hid) REFERENCES hausmeister(hid))
    """ 
    cursor.execute(sql)

    # ---- Tabelle trainingseinheit  --------------
    sql = """
    CREATE TABLE trainingseinheit (tag TEXT, zeit TEXT, uid TEXT, ort TEXT,
    PRIMARY KEY(tag,zeit,uid),
    FOREIGN KEY(ort) REFERENCES trainingsort(ort),
    FOREIGN KEY(uid) REFERENCES uebungsgruppe(uid))
    """ 
    cursor.execute(sql)

    # ---- Tabelle hausmeister  --------------
    sql = """
    CREATE TABLE hausmeister (hid INTEGER PRIMARY KEY, name TEXT, telefon Text)
    """ 
    cursor.execute(sql)



    connection.close()
    
def mitglieder_fuellen():
    # ---- Tabelle mitglieder  
    connection = sq.connect("sportverein.db")
    cursor = connection.cursor()

    datalist = [[1963, 'Giek', 'Ulrike', '0641-3062542', '83332789', 93750115, 52.00], \
    [1964, 'Kiefer', 'Claus', '0341-4112152', '74197742', 56250069, 52.00], \
    [1965, 'Voss', 'Heinz Gunter', '', '41465530', 18750023, 17.00], \
    [1966, 'Matthies', 'Werner', '', '11167490', 18750023, 34.00], \
    [1967, 'Krey', 'Manfred', '', '92011841', 37500046, 17.00], \
    [1968, 'Rehbock', 'Helga', '', '43016556', 37500046, 34.00], \
    [1969, 'Siebert', 'Ernst', '', '79257785', 18750023, 34.00], \
    [1970, 'Auffarth', 'Werner', '0561-982660', '85603569', 93750115, 34.00], \
    [1971, 'Klein', 'Gerhard', '06451-4074', '39197078', 75000092, 34.00], \
    [1972, 'Bartussek', 'Werner', '', '85293740', 18750023, 34.00], \
    [1973, 'Beck', 'Werner', '', '31894116', 37500046, 34.00], \
    [1974, 'Brinkmann', 'Werner', '06655-2463', '44416647', 56250069, 17.00], \
    [1975, 'Kinner', 'Friedemann', '06650-578', '10592491', 37500046, 52.00], \
    [1976, 'Krummheuer', 'Werner', '0551-371667', '28000905', 18750023, 52.00], \
    [1977, 'Schinkmann', 'Werner', '', '11218249', 93750115, 52.00], \
    [1978, 'Mahlmann', 'Ursula', '', '45124439', 93750115, 17.00], \
    [1979, 'Oehring', 'Werner', '', '6459323', 75000092, 34.00], \
    [1980, 'Binsfeld', 'Berthold', '02323-163245', '61843123', 75000092, 17.00], \
    [1981, 'Heinrich', 'Werner', '', '97368872', 56250069, 52.00], \
    [1982, 'Seiler', 'Werner', '', '78634467', 56250069, 17.00], \
    [1983, 'Loewen', 'Werner', '', '9503515', 18750023, 17.00], \
    [1984, 'Hagen', 'Karl-Heinz', '', '25580114', 93750115, 17.00], \
    [1985, 'Schaepers', 'Werner', '', '24992406', 37500046, 34.00], \
    [1986, 'Bellenbaum', 'Peter', '', '47381070', 18750023, 52.00], \
    [1987, 'Domsel', 'Reinhold', '02064-54050', '33653645', 18750023, 34.00], \
    [1988, 'Deest', 'Erna', '02064-66224', '16824824', 93750115, 17.00], \
    [1989, 'Hipp', 'Werner', '', '74820069', 93750115, 34.00], \
    [1990, 'Rienhöfer', 'Franz-Josef', '', '915766', 37500046, 52.00], \
    [1991, 'Budde', 'Werner', '02951-98980', '91293957', 56250069, 17.00], \
    [1992, 'Holthaus', 'Werner', '', '77550737', 56250069, 52.00], \
    [1993, 'Ellerbusch', 'Dietrich', '', '95238948', 75000092, 52.00], \
    [1994, 'Holling', 'Ernst Friedrich', '', '80999916', 37500046, 34.00], \
    [1995, 'Heuer', 'Leo', '', '65598242', 56250069, 52.00], \
    [1996, 'Brandt', 'Werner', '05772-97100', '9661379', 93750115, 17.00], \
    [1997, 'Schotten', 'Herbert', '', '78797557', 56250069, 34.00], \
    [1998, 'Homberg', 'Ernst', '', '50550821', 37500046, 17.00], \
    [1999, 'Strohmayer', 'Uta', '', '7658408', 75000092, 34.00], \
    [2000, 'Fischer', 'Berthold', '', '14877014', 93750115, 17.00], \
    [2001, 'Kirchner', 'Werner', '0611-97897854', '37979923', 75000092, 17.00], \
    [2002, 'Winterlich', 'Helmut', '0241-1769200', '28577155', 18750023, 17.00], \
    [2004, 'Hager', 'Werner', '', '18959486', 37500046, 17.00], \
    [2005, 'Becker', 'Reinhard', '', '52093407', 56250069, 17.00], \
    [2007, 'Hahne', 'Josef', '02307-982320', '20289753', 75000092, 52.00], \
    [2008, 'Eggenkaemper', 'Christian', '', '9892999', 93750115, 52.00], \
    [2009, 'Farwik', 'Werner', '', '71646932', 56250069, 52.00], \
    [2010, 'Böttcher', 'Wilhelm', '', '69987472', 56250069, 34.00], \
    [2011, 'Senkler', 'Anton', '', '79447266', 37500046, 34.00], \
    [2012, 'Hellwig', 'Goetz', '', '70200962', 56250069, 52.00], \
    [2013, 'Möller', 'Herbert', '069-80652145', '67573040', 18750023, 34.00], \
    [2014, 'Fried', 'Werner', '', '87364', 93750115, 52.00], \
    [2015, 'Bagel', 'Werner', '', '68339572', 56250069, 34.00], \
    [2016, 'Heil', 'Werner', '', '83199807', 56250069, 17.00], \
    [2017, 'Hüttenberger', 'Gerd', '', '39160427', 56250069, 52.00], \
    [2018, 'Schäfer', 'Volker', '06151-132566', '69310738', 75000092, 17.00], \
    [2019, 'König', 'Christiane', '06062-3691', '5586741', 37500046, 34.00], \
    [2020, 'Ohlweiler', 'Werner', '', '96367432', 56250069, 34.00], \
    [2021, 'Maurer', 'Heidrun', '', '47444311', 37500046, 52.00], \
    [2022, 'Steitz', 'Dietrich', '0611-312265', '96536419', 93750115, 52.00], \
    [2023, 'Fluhme', 'Werner', '', '87603192', 56250069, 17.00], \
    [2024, 'Sarges', 'Claus', '0641-3062569', '46856200', 18750023, 17.00], \
    [2025, 'Huhn', 'Werner', '06172-690026', '12167013', 75000092, 52.00], \
    [2026, 'Fey', 'Dieter', '', '90995706', 37500046, 34.00], \
    [2027, 'Richter', 'Roland', '', '64831663', 75000092, 17.00], \
    [2028, 'Richter', 'Simon', '06078-939311', '25773343', 75000092, 17.00], \
    [2029, 'Rexroth', 'Werner', '06071-41112', '72581897', 56250069, 52.00], \
    [2030, 'Regh', 'Werner', '0611-312308', '83952460', 18750023, 17.00], \
    [2031, 'Winkelmann', 'Klaus', '0621-735125', '89004021', 37500046, 34.00], \
    [2032, 'Roos', 'Frauke', '', '90587092', 56250069, 34.00], \
    [2033, 'Wagner', 'Dieter', '', '97266989', 93750115, 17.00], \
    [2034, 'Schulte', 'Franz', '0721-978370', '66286702', 56250069, 34.00], \
    [2035, 'Schlauderer', 'Josef', '', '80606316', 37500046, 34.00], \
    [2036, 'Zenk', 'Klaus', '', '50365887', 56250069, 17.00], \
    [2037, 'Nusime', 'Viktor', '0043-7722-3125', '99463598', 75000092, 34.00], \
    [2038, 'Verheggen', 'Werner', '', '84696463', 75000092, 17.00], \
    [2039, 'Poecking', 'Georg', '', '93462631', 37500046, 52.00], \
    [2040, 'Steffens', 'Guenther', '', '26449562', 56250069, 52.00], \
    [2041, 'Kerski', 'Manfred', '', '37800545', 18750023, 34.00], \
    [2042, 'Wiedemann', 'Bernd', '', '13760322', 56250069, 17.00], \
    [2043, 'Zeglin', 'Emil', '', '69735982', 56250069, 34.00], \
    [2044, 'Weyand', 'Helmut', '', '52662368', 75000092, 34.00], \
    [2045, 'Vogt', 'Gerhard', '02541-5221', '96419818', 75000092, 17.00], \
    [2046, 'Tegethoff', 'Werner', '', '27896765', 18750023, 34.00], \
    [2047, 'Thöle', 'Juergen', '', '60934507', 56250069, 34.00], \
    [2048, 'Hüsch', 'Hans-Peter', '', '26260553', 75000092, 34.00], \
    [2049, 'Fisbach', 'Alfred', '', '61127351', 37500046, 17.00], \
    [2050, 'Dasbach', 'Bernfried', '', '67394855', 18750023, 34.00], \
    [2051, 'Schmitz', 'Werner', '', '5331080', 18750023, 34.00], \
    [2052, 'Laubersheimer', 'Joachim', '', '99475484', 93750115, 17.00], \
    [2053, 'Loehring', 'Werner', '', '84464538', 75000092, 52.00], \
    [2054, 'Hahn', 'Wolfgang', '', '95504985', 18750023, 52.00], \
    [2055, 'Preckwinkel', 'Barbara', '', '52298590', 18750023, 17.00], \
    [2056, 'Niggemeier', 'Georg', '', '82417891', 37500046, 52.00], \
    [2057, 'Jonas', 'Volker', '0541-17441', '97495267', 56250069, 34.00], \
    [2058, 'Grobelny', 'Uwe', '', '2967007', 56250069, 34.00], \
    [2059, 'Mueller', 'Günter', '', '66491242', 56250069, 34.00], \
    [2060, 'Guminski', 'Dirk', '', '18317882', 56250069, 52.00], \
    [2061, 'Lorenz', 'Arnold', '', '68655744', 93750115, 34.00], \
    [2062, 'Beisner', 'Harry', '', '97103396', 37500046, 17.00], \
    [2063, 'Buse', 'Franz', '', '16788482', 56250069, 52.00], \
    [2064, 'Deutsch', 'Albert', '', '64024385', 18750023, 34.00], \
    [2065, 'Pieper', 'Gerhard', '02173-955910', '57854373', 75000092, 52.00], \
    [2066, 'Jakobs', 'Wolfgang', '', '58452294', 93750115, 34.00], \
    [2067, 'Mader', 'Werner', '', '7709995', 18750023, 17.00], \
    [2068, 'Heuser', 'Werner', '02841-98366', '81899418', 75000092, 52.00], \
    [2069, 'Thomas', 'Alfred', '', '83533557', 56250069, 17.00], \
    [2070, 'Boerste', 'Juergen', '', '48501630', 18750023, 52.00], \
    [2071, 'Fritzowski', 'Wolfgang', '', '66903409', 93750115, 52.00], \
    [2072, 'Weingartner', 'Werner', '', '45607929', 93750115, 34.00], \
    [2073, 'Doehl', 'Joerg', '', '32465042', 56250069, 52.00], \
    [2074, 'Menzel', 'Werner', '', '39974244', 93750115, 34.00], \
    [2075, 'Vorbrook', 'Heinz', '', '47126803', 75000092, 17.00], \
    [2076, 'Hoffmann', 'Werner', '', '60183672', 56250069, 17.00], \
    [2077, 'Grosche', 'Werner', '', '6983628', 75000092, 52.00], \
    [2078, 'Zang', 'Heinz', '', '73836369', 93750115, 34.00], \
    [2079, 'Gaehrken', 'Clemens', '', '67860972', 56250069, 17.00], \
    [2080, 'Wehling', 'Werner', '', '39367051', 37500046, 52.00], \
    [2081, 'Brockhoff', 'Hans', '', '42035915', 18750023, 34.00], \
    [2082, 'Nassenstein', 'Theo', '', '97633194', 93750115, 34.00], \
    [2083, 'Rathje', 'Werner', '', '21097667', 56250069, 52.00], \
    [2084, 'Montag', 'Theodor', '', '88496529', 93750115, 34.00], \
    [2085, 'Keller', 'Werner', '0203-555940', '61164908', 75000092, 34.00], \
    [2086, 'Steigerwald', 'Werner', '', '98615797', 56250069, 34.00], \
    [2087, 'Toernig', 'Werner', '', '35711970', 18750023, 52.00], \
    [2088, 'Ballmann', 'Werner', '', '11087012', 56250069, 52.00], \
    [2089, 'Kuhn', 'Willibald', '', '39419574', 93750115, 34.00], \
    [2090, 'Pogadel', 'Fritz', '', '24780713', 75000092, 34.00], \
    [2091, 'Schulz', 'Wolfgang', '', '89449429', 93750115, 52.00], \
    [2092, 'Jöhren', 'Werner', '', '76553899', 93750115, 52.00], \
    [2093, 'Fritzen', 'Alfred', '', '41391320', 18750023, 34.00], \
    [2094, 'Viehweger', 'Horst', '', '63916588', 75000092, 17.00], \
    [2095, 'Schreieck', 'Horst', '', '46096847', 56250069, 52.00], \
    [2096, 'Knuelle', 'Werner', '', '49723880', 18750023, 34.00], \
    [2097, 'Hepelmann', 'Siegfried', '', '9253778', 75000092, 34.00], \
    [2098, 'Niederkleiner', 'Werner', '', '28259931', 37500046, 34.00], \
    [2099, 'Lotter', 'Klaus', '0201-510315', '64676064', 93750115, 34.00], \
    [2100, 'Meissner', 'Juergen', '06155-87540', '10621795', 18750023, 17.00], \
    [2101, 'Diez', 'Norbert', '06126-2024', '20517453', 75000092, 52.00], \
    [2102, 'Schwalbenbach', 'Werner', '0611-313671', '36841214', 93750115, 17.00], \
    [2103, 'Simon', 'Ulrich', '', '90292302', 56250069, 17.00], \
    [2104, 'Gassel', 'Ralf', '', '42721318', 75000092, 17.00], \
    [2105, 'Leifeld', 'Werner', '', '17200365', 18750023, 52.00], \
    [2106, 'Widulla', 'Kurt', '', '86548173', 75000092, 52.00], \
    [2107, 'Hörsken', 'Werner', '02841-790614', '15150692', 56250069, 52.00], \
    [2108, 'Frehn', 'Werner', '', '37681925', 18750023, 34.00], \
    [2109, 'Oberdörster', 'Werner', '02161-89196', '20117975', 18750023, 34.00], \
    [2110, 'Nevries', 'Fritz', '', '22811121', 37500046, 34.00], \
    [2111, 'Oselies', 'Klaus-Henning', '', '86518338', 75000092, 52.00], \
    [2112, 'Wernicke', 'Werner', '', '13714430', 18750023, 34.00], \
    [2113, 'Rimpler', 'Gerhard', '02103-3636', '98441813', 75000092, 34.00], \
    [2114, 'Schuetz', 'Dieter', '', '11564407', 56250069, 34.00], \
    [2115, 'Lohmann', 'Gernot', '', '62993772', 56250069, 17.00], \
    [2116, 'Hahn', 'Werner', '05254-2745', '36340617', 56250069, 17.00], \
    [2117, 'Wanner', 'Werner', '', '76750471', 37500046, 52.00], \
    [2118, 'Minkler', 'Jürgen', '', '45176490', 18750023, 52.00], \
    [2119, 'Schaedel', 'Christoph', '', '81268182', 56250069, 17.00], \
    [2120, 'Bertermann', 'Werner', '', '74342180', 18750023, 34.00], \
    [2121, 'Körner', 'Werner', '', '69028398', 75000092, 52.00], \
    [2122, 'Menges', 'Waldemar', '', '60295488', 75000092, 52.00], \
    [2123, 'Seiler', 'Klaus', '', '86276458', 75000092, 17.00], \
    [2124, 'Breitschaft', 'Bernd', '', '48446228', 75000092, 34.00], \
    [2125, 'Standt', 'Bernd', '', '35763475', 18750023, 17.00], \
    [2126, 'Holtorf', 'Wolfgang', '', '76724932', 56250069, 34.00], \
    [2127, 'Jacobs', 'Hans-Joachim', '', '15938485', 18750023, 17.00], \
    [2128, 'Schmitz', 'Werner', '', '78123691', 37500046, 52.00], \
    [2129, 'Hentschke', 'Axel', '06173-63018', '24885861', 37500046, 34.00], \
    [2130, 'Thomas', 'Ulf', '02241-805422', '56678043', 75000092, 52.00], \
    [2131, 'Berger', 'Peter-Henning', '', '87882894', 93750115, 34.00], \
    [2132, 'Droste', 'Horst', '02921-52478', '47806376', 37500046, 17.00], \
    [2133, 'Steinmetz', 'Werner', '05141-44821', '81765483', 75000092, 52.00], \
    [2134, 'Wolter', 'Werner', '', '32480360', 18750023, 34.00], \
    [2135, 'Liebethal', 'Juergen', '', '97417182', 56250069, 17.00], \
    [2136, 'Nordsiek', 'Joachim', '', '90591321', 37500046, 34.00], \
    [2137, 'Weber', 'Ernst', '05971-51016', '68333552', 56250069, 52.00], \
    [2138, 'Scharff', 'Rudi', '', '82018157', 56250069, 52.00], \
    [2139, 'Rost', 'Werner', '', '79173976', 37500046, 52.00], \
    [2140, 'Ruholl', 'Ulrich', '', '77023804', 93750115, 52.00], \
    [2141, 'Zehfuss', 'Werner', '0531-2621036', '39086195', 18750023, 34.00], \
    [2142, 'Kriep', 'Werner', '', '21691997', 56250069, 52.00], \
    [2143, 'Bohnes', 'Werner', '', '77395705', 18750023, 17.00], \
    [2144, 'Buss', 'Werner', '', '89624613', 75000092, 52.00], \
    [2145, 'Lenarz', 'Richard', '0221-501582', '20884103', 75000092, 34.00], \
    [2146, 'Trieschnigg', 'Werner', '', '99132623', 75000092, 34.00], \
    [2147, 'Zickel', 'Raimund', '', '22000421', 37500046, 17.00], \
    [2148, 'Elle', 'Werner', '', '61087293', 75000092, 52.00], \
    [2149, 'Kriek', 'Joachim', '', '94148657', 56250069, 52.00], \
    [2150, 'Happ', 'Manfred', '', '56489198', 75000092, 34.00], \
    [2151, 'Schuetz', 'Peter', '', '13965365', 18750023, 34.00], \
    [2152, 'Lotze', 'Werner', '', '15896114', 93750115, 17.00], \
    [2153, 'Posthaus', 'Werner', '', '65994526', 37500046, 52.00], \
    [2154, 'Kuntze', 'Horst', '', '70004926', 37500046, 52.00], \
    [2155, 'Hill', 'Werner', '', '72806195', 18750023, 34.00], \
    [2156, 'Görner', 'Uwe', '06181-98470', '53729047', 56250069, 34.00], \
    [2157, 'Kessler', 'Johannes', '', '24107005', 18750023, 17.00], \
    [2158, 'Haarmann', 'Werner', '', '66442304', 93750115, 17.00], \
    [2159, 'Fischer', 'Burkhard', '', '95806388', 75000092, 34.00], \
    [2160, 'Fokken', 'Hermann', '', '17964617', 56250069, 52.00], \
    [2161, 'Steiner', 'Heinz', '06162-93130', '31327841', 37500046, 17.00], \
    [2162, 'Mahn', 'Karl-Heinz', '02602-3188', '27039000', 18750023, 52.00], \
    [2163, 'Schmidtmer', 'Werner', '06151-132560', '44485277', 56250069, 52.00], \
    [2164, 'Linnebur', 'Johannes', '02941-980224', '64063141', 37500046, 52.00], \
    [2165, 'Rasmus', 'Werner', '', '9215694', 93750115, 34.00], \
    [2166, 'Keil', 'Heinz', '0661-54537', '88177191', 37500046, 34.00], \
    [2167, 'Ressel', 'Werner', '', '94057660', 18750023, 17.00], \
    [2168, 'Thoma', 'Werner', '', '27103945', 18750023, 17.00], \
    [2169, 'Schilling', 'Klaus', '', '36018041', 37500046, 17.00], \
    [2170, 'Kaestle', 'Werner', '', '55222565', 18750023, 34.00], \
    [2171, 'Haschberger', 'Frau', '', '25654784', 93750115, 17.00], \
    [2172, 'Hartwich', 'Werner', '06196-95700', '79966370', 18750023, 34.00], \
    [2173, 'Weinert', 'Werner', '', '94785403', 75000092, 17.00], \
    [2174, 'Friedrich', 'Werner', '', '58278263', 75000092, 34.00], \
    [2175, 'Sommer', 'Eberhard', '', '82779013', 75000092, 17.00], \
    [2176, 'Wehling', 'Reinhold van', '', '66873586', 37500046, 52.00], \
    [2177, 'Kaehler', 'Werner', '', '44361848', 75000092, 34.00], \
    [2178, 'Schmidt', 'Werner', '', '89593183', 75000092, 17.00], \
    [2179, 'Deest', 'Werner', '', '92630005', 18750023, 17.00], \
    [2180, 'Warntjen', 'Bernhard', '', '32507660', 18750023, 34.00], \
    [2181, 'Hildebrand', 'Werner', '', '56249820', 18750023, 52.00], \
    [2182, 'Ingenlath', 'Werner', '', '90727896', 37500046, 34.00], \
    [2183, 'Patte', 'Werner', '02553-93520', '63755219', 93750115, 17.00], \
    [2184, 'Niegemann', 'Gerhard', '', '33977367', 56250069, 17.00], \
    [2185, 'Schäfer', 'Helmut', '', '57618121', 37500046, 52.00], \
    [2186, 'Sasse', 'Siegfried', '', '87761068', 75000092, 17.00], \
    [2187, 'Jung', 'Rolf', '', '278694', 93750115, 34.00], \
    [2188, 'Frenzl', 'Joachim', '06452-3665', '84015477', 56250069, 17.00], \
    [2189, 'Roth', 'Werner', '', '48132485', 18750023, 34.00], \
    [2190, 'Hartleif', 'Werner', '', '62688488', 75000092, 34.00], \
    [2191, 'Paetz', 'Werner', '', '5430063', 75000092, 52.00], \
    [2192, 'Bund', 'Werner', '', '63069665', 18750023, 34.00], \
    [2193, 'Beckmann', 'Werner', '', '11864278', 75000092, 17.00], \
    [2194, 'Kluska', 'Werner', '', '2925132', 56250069, 17.00], \
    [2195, 'Mueller', 'Werner', '', '62695595', 18750023, 34.00], \
    [2196, 'Kuhn', 'Herbert', '', '24873762', 56250069, 17.00], \
    [2197, 'Pollmann', 'Michael', '', '53330969', 93750115, 17.00], \
    [2198, 'Dreiner', 'Werner', '', '33610022', 56250069, 52.00], \
    [2199, 'Hein', 'Werner', '', '83643508', 75000092, 17.00], \
    [2200, 'Kluge', 'Horst', '', '96442808', 75000092, 34.00], \
    [2201, 'Eßt', 'Werner', '', '14287746', 37500046, 34.00], \
    [2202, 'Heese', 'Heidrun', '', '34150250', 18750023, 34.00], \
    [2203, 'Bernstein', 'Werner', '', '63259451', 37500046, 17.00], \
    [2204, 'Falk', 'Werner', '', '1600464', 93750115, 34.00], \
    [2205, 'Sarges', 'Werner', '', '99012788', 75000092, 34.00], \
    [2206, 'Schäfers', 'Werner', '', '7943773', 75000092, 17.00], \
    [2207, 'Kleissner', 'Ernst', '', '10995839', 93750115, 17.00], \
    [2208, 'Luebbers', 'Werner', '', '7082214', 93750115, 34.00], \
    [2209, 'Koeper', 'Klaus-Georg', '', '30779701', 37500046, 52.00], \
    [2210, 'Gerz', 'Werner', '', '14448347', 56250069, 52.00], \
    [2211, 'Kaul', 'Werner', '0561-13871', '33384396', 37500046, 34.00], \
    [2212, 'Schindler', 'Ralf', '', '91550811', 37500046, 34.00], \
    [2213, 'Klein', 'Juergen', '', '94490713', 75000092, 17.00], \
    [2214, 'Becker', 'Willy', '', '5416827', 56250069, 17.00], \
    [2215, 'Wittkopp', 'Werner', '', '40624684', 93750115, 52.00], \
    [2216, 'Rau', 'Werner', '', '79080779', 93750115, 17.00], \
    [2217, 'Gerber', 'Karl', '02752-1599', '10655043', 75000092, 52.00], \
    [2218, 'Hinz', 'Wolfgang', '', '79321001', 37500046, 17.00], \
    [2219, 'Bertrand', 'Gerd', '', '48068415', 18750023, 52.00], \
    [2220, 'Niemann', 'Heinz-Gerd', '', '85857559', 93750115, 52.00], \
    [2221, 'Kohlitz', 'Werner', '', '8005665', 18750023, 52.00], \
    [2222, 'Carpani', 'Gerhard', '', '51228318', 75000092, 52.00], \
    [2223, 'Lindner', 'Ursula', '', '94347917', 56250069, 52.00], \
    [2224, 'Muckenhaupt', 'Michael', '', '5452868', 75000092, 34.00], \
    [2225, 'Krug', 'Hubertus', '', '46352794', 93750115, 52.00], \
    [2226, 'Großefrie-Beckers', 'Bertram', '02323-163295', '51424350', 18750023, 52.00], \
    [2227, 'Auth', 'Klaus', '', '89050586', 37500046, 17.00], \
    [2228, 'Hoefling', 'Klaus', '', '73201322', 56250069, 52.00], \
    [2229, 'Rudolph', 'Peter', '', '94621720', 18750023, 17.00], \
    [2230, 'Pastewka', 'Reinold', '', '7250771', 37500046, 17.00], \
    [2231, 'Megharbi', 'Wolfgang', '', '63713279', 93750115, 17.00], \
    [2232, 'Braun', 'Werner', '', '29082115', 18750023, 17.00], \
    [2233, 'Howoritsch', 'Werner', '', '83114766', 37500046, 34.00], \
    [2234, 'Kuhn', 'Hans-Juergen', '0561-18265', '9888525', 18750023, 52.00], \
    [2235, 'Reinhardt', 'Guenter', '', '96441784', 93750115, 52.00], \
    [2236, 'Bamberger', 'Friedrich', '', '97073647', 56250069, 34.00], \
    [2237, 'Franke', 'Werner', '', '16785229', 93750115, 52.00], \
    [2238, 'Terhorst', 'Werner', '', '9353283', 75000092, 52.00], \
    [2239, 'Kranenberg', 'Werner', '', '635299', 93750115, 52.00], \
    [2240, 'Klunker', 'Ulrich', '', '77474100', 18750023, 52.00], \
    [2241, 'Spangenberg', 'Werner', '', '7213444', 93750115, 52.00], \
    [2242, 'Oberdick', 'Helge', '', '36362476', 37500046, 52.00], \
    [2243, 'Manfraß', 'Werner', '', '44137462', 56250069, 34.00], \
    [2244, 'vom Stein', 'Wolfgang', '02196-6018', '18335064', 93750115, 34.00], \
    [2245, 'Lehmann', 'Margrit', '06223-9610', '57423726', 37500046, 34.00], \
    [2246, 'Böhler', 'Werner', '', '20564957', 56250069, 52.00], \
    [2247, 'Heuser', 'Werner', '06659-1894', '35240448', 75000092, 52.00], \
    [2248, 'Mizgalski', 'Guenter', '', '86211939', 93750115, 34.00], \
    [2249, 'Rumpenhorst', 'Otmar', '', '64059722', 56250069, 52.00], \
    [2250, 'Müller', 'Erich', '', '51758404', 93750115, 52.00], \
    [2251, 'Dionisius', 'Kurt', '', '3505538', 75000092, 34.00], \
    [2252, 'Scholl', 'Wolfgang', '', '17584765', 18750023, 17.00], \
    [2253, 'Jännert', 'Horst', '', '47160429', 93750115, 34.00], \
    [2257, 'Görner', 'Werner', '0541-3232026', '94647481', 37500046, 52.00], \
    [2258, 'Mees', 'Joachim', '', '40221954', 93750115, 34.00], \
    [2259, 'Ihlemann', 'Kurt', '', '10382714', 93750115, 17.00], \
    [2260, 'Thieme', 'Werner', '', '2277121', 37500046, 52.00], \
    [2261, 'Hänsel', 'Werner', '', '71578049', 93750115, 17.00], \
    [2262, 'Richter', 'Werner', '0661-66028', '12273364', 93750115, 52.00], \
    [2263, 'Stoll', 'Claus', '0611-317110', '78443745', 18750023, 52.00], \
    [2264, 'Berndt', 'Klaus', '06629-6555', '4036326', 37500046, 34.00], \
    [2265, 'Romeis', 'Gisela', '', '38538290', 18750023, 34.00], \
    [2266, 'Rühenbeck', 'Burkhard', '', '13046617', 56250069, 34.00], \
    [2267, 'Mombaur', 'Frau', '', '74493827', 56250069, 52.00], \
    [2268, 'Grieße', 'Werner', '', '17755202', 56250069, 52.00], \
    [2269, 'Boettcher', 'Werner', '', '11690625', 18750023, 52.00], \
    [2270, 'Heinrich', 'Rochus', '06151-132552', '84336639', 93750115, 34.00], \
    [2271, 'Lünstedt', 'Bernhard', '', '45893273', 75000092, 34.00], \
    [2272, 'Hildebrandt', 'Wolfram', '', '28255160', 18750023, 34.00], \
    [2273, 'Macke', 'Werner', '04208-594', '84489433', 75000092, 17.00], \
    [2274, 'Patte', 'Christoph', '', '13905817', 56250069, 34.00], \
    [2275, 'Meibaum', 'Berthold', '', '15067839', 93750115, 34.00], \
    [2276, 'Brauer', 'Werner', '', '45184566', 37500046, 34.00], \
    [2277, 'Kramer', 'Wolfram', '06631-2229', '17007948', 75000092, 52.00], \
    [2278, 'Heinrich', 'Peter', '', '52585956', 37500046, 34.00], \
    [2279, 'Euler', 'Hartmut', '', '12176017', 56250069, 34.00], \
    [2280, 'Kurtz', 'Werner', '0711-7802228', '42318555', 18750023, 52.00], \
    [2281, 'Klein', 'Wolfgang', '0621-504430110', '47965411', 37500046, 52.00], \
    [2282, 'Schmidt', 'Erika', '', '54677984', 93750115, 17.00], \
    [2283, 'Marx', 'Werner', '', '72882637', 56250069, 34.00], \
    [2284, 'Voß', 'Klaus', '', '38487824', 93750115, 17.00], \
    [2285, 'Mees', 'Werner', '', '64871674', 93750115, 52.00], \
    [2286, 'Walter', 'Christian', '', '98241599', 93750115, 17.00], \
    [2287, 'Wehling', 'Hermann-Josef', '', '46567638', 75000092, 52.00], \
    [2288, 'Aulich', 'Werner', '', '62293553', 75000092, 34.00], \
    [2289, 'Nürnberger', 'Werner', '', '67749332', 56250069, 52.00], \
    [2290, 'Vilmar', 'Udo', '06151-132544', '63034153', 75000092, 17.00], \
    [2291, 'Dahmen', 'Bernd', '02251-14250', '15017532', 56250069, 17.00], \
    [2292, 'Herling', 'Werner', '', '99663546', 37500046, 34.00], \
    [2293, 'Joppich', 'Thomas', '', '45219680', 18750023, 34.00], \
    [2294, 'Fuchs', 'Karl', '', '7163148', 75000092, 52.00], \
    [2295, 'Meyer', 'Werner', '', '91038680', 18750023, 34.00], \
    [2296, 'Wagner', 'Werner', '', '87101455', 18750023, 17.00], \
    [2297, 'Geise', 'Werner', '', '14166375', 18750023, 52.00], \
    [2298, 'Hruza', 'Werner', '', '94269713', 75000092, 34.00], \
    [2299, 'Friese', 'Werner', '', '91082240', 18750023, 34.00], \
    [2300, 'Hüttenberg', 'Bernd', '', '19212907', 56250069, 34.00], \
    [2301, 'Krause', 'Hans-Ludwig', '', '11327997', 56250069, 17.00], \
    [2302, 'Rzytki', 'Werner', '0611-312249', '89650190', 18750023, 34.00], \
    [2303, 'Kolbe', 'Klaus', '', '50752913', 75000092, 34.00], \
    [2304, 'Krebs', 'Ulrich', '', '4222838', 75000092, 34.00], \
    [2305, 'Mischitz', 'Reinhold', '06654-7081', '73245709', 93750115, 52.00], \
    [2306, 'Ancke', 'Werner', '', '40638281', 37500046, 17.00], \
    [2307, 'Schelper', 'Werner', '05602-3262', '7600587', 56250069, 17.00], \
    [2308, 'Janotte', 'Rudolf', '', '43069699', 93750115, 52.00], \
    [2309, 'Schmidt', 'Bernd', '', '72442177', 56250069, 34.00], \
    [2310, 'Loewen', 'Helmut', '', '35675408', 75000092, 34.00], \
    [2311, 'Rohde', 'Werner', '', '96592586', 37500046, 17.00], \
    [2312, 'Stängle', 'Werner', '', '31599504', 93750115, 34.00], \
    [2313, 'Bethmann', 'Werner', '', '92863780', 18750023, 17.00], \
    [2314, 'Grupe', 'Ute', '', '61556340', 18750023, 52.00], \
    [2315, 'Struss', 'Heinz', '', '77417969', 93750115, 17.00], \
    [2316, 'Haas', 'Christian', '', '63813502', 56250069, 34.00], \
    [2317, 'Schurau', 'Peter', '', '13534353', 75000092, 52.00], \
    [2318, 'Heß', 'Werner', '', '71873686', 37500046, 52.00], \
    [2319, 'Meier', 'Werner', '', '81052853', 75000092, 34.00], \
    [2320, 'Hense', 'Werner', '', '55305378', 75000092, 52.00], \
    [2321, 'Kaufhold', 'Werner', '', '17493945', 18750023, 52.00], \
    [2322, 'Mader', 'Werner', '', '20748414', 93750115, 34.00], \
    [2323, 'Winter', 'Thomas', '030-25887911', '18602971', 37500046, 52.00], \
    [2324, 'Schirmeisen', 'Gerd', '', '60144490', 18750023, 17.00], \
    [2325, 'Zimmermann', 'Werner', '06433-2076', '48450222', 56250069, 17.00], \
    [2326, 'Plasche', 'Karl', '', '2882417', 56250069, 52.00], \
    [2327, 'Gelueck', 'Werner', '', '44796048', 75000092, 52.00], \
    [2328, 'Ballendatt', 'Rudolf', '', '87181629', 93750115, 34.00], \
    [2329, 'Werner', 'Werner', '', '61655067', 56250069, 17.00], \
    [2330, 'Müller', 'Werner', '06103-73489', '36713908', 75000092, 17.00], \
    [2331, 'Meyer', 'Bärbel', '', '50659993', 75000092, 17.00], \
    [2332, 'Helmold', 'Hartmut', '', '42511549', 93750115, 52.00], \
    [2333, 'Sick', 'Otto', '', '91618238', 75000092, 34.00], \
    [2334, 'Uerkvitz', 'Werner', '', '27741938', 75000092, 34.00], \
    [2335, 'Schrobang', 'Karl', '', '58755948', 75000092, 52.00], \
    [2336, 'Schäfer', 'Werner', '', '11233735', 18750023, 17.00], \
    [2337, 'Küper', 'Dieter', '', '5398175', 18750023, 34.00], \
    [2338, 'Plesse', 'Peter', '', '27141812', 56250069, 52.00], \
    [2339, 'Kraus', 'Werner', '', '72438041', 37500046, 17.00], \
    [2340, 'Nikolayczik', 'Werner', '', '66156118', 75000092, 17.00], \
    [2341, 'Hoffmann', 'Günter', '', '85881847', 56250069, 34.00], \
    [2342, 'Weitkamp', 'Werner', '', '16204042', 56250069, 34.00], \
    [2343, '…hlmann', 'Werner', '040-3397', '41336943', 75000092, 52.00], \
    [2344, 'Schürg', 'Werner', '0271-356969', '50016579', 93750115, 34.00], \
    [2345, 'Neitzel', 'Werner', '', '28653572', 56250069, 52.00], \
    [2346, 'Hungenberg', 'Werner', '', '80601053', 75000092, 34.00], \
    [2347, 'Pane', 'Werner', '', '61913826', 37500046, 34.00], \
    [2348, 'Nikl', 'Werner', '', '85689561', 37500046, 34.00], \
    [2349, 'Zwick', 'Werner', '', '84466079', 93750115, 17.00], \
    [2350, 'Thie', 'Dirk', '', '21401659', 93750115, 52.00], \
    [2352, 'Fokken', 'Werner', '', '4868502', 56250069, 17.00], \
    [2353, 'Schweikart', 'Werner', '06083-835', '24916789', 93750115, 52.00], \
    [2354, 'Will', 'Werner', '', '76478880', 18750023, 52.00], \
    [2355, 'Lohwasser', 'Werner', '', '80542883', 75000092, 34.00], \
    [2356, 'Hamm', 'Werner', '', '84248474', 93750115, 17.00], \
    [2357, 'Maser', 'Ulrich', '', '64117390', 18750023, 17.00], \
    [2358, 'Brettschneider', 'Werner', '069-21238752', '20980669', 93750115, 52.00], \
    [2359, 'Gerland', 'Werner', '', '22111861', 37500046, 52.00], \
    [2360, 'Buchmann', 'Claudia', '', '34048592', 56250069, 52.00], \
    [2361, 'Böttcher', 'Werner', '', '54687908', 75000092, 34.00], \
    [2362, 'Stimpel', 'Werner', '', '39686045', 18750023, 34.00], \
    [2363, 'Auer', 'Werner', '', '3366520', 18750023, 17.00], \
    [2364, 'Shripczynski', 'Norbert', '06151-13265', '81103617', 56250069, 17.00], \
    [2365, 'Wernicke', 'Werner', '', '8503284', 93750115, 34.00], \
    [2366, 'Sahler', 'Werner', '', '14698603', 75000092, 17.00], \
    [2367, 'Schulz', 'Werner', '', '39431763', 75000092, 52.00], \
    [2368, 'Müller', 'Werner', '', '29654712', 56250069, 17.00], \
    [2369, 'Pohl', 'Georg', '', '6744767', 56250069, 52.00], \
    [2370, 'Lotz', 'Werner', '', '59315127', 56250069, 17.00], \
    [2371, 'Dege', 'Rudolf', '', '9354732', 56250069, 17.00], \
    [2372, 'Hesse', 'Rudi', '', '24997221', 37500046, 34.00], \
    [2373, 'Behlau', 'Günter', '', '28301032', 56250069, 34.00], \
    [2375, 'Trosien', 'Werner', '0081-45-9414841', '22493127', 56250069, 17.00], \
    [2376, 'Bubeck', 'Emil', '07156-205280', '41994079', 93750115, 52.00], \
    [2377, 'Henne', 'Werner', '07082-5629', '94490266', 37500046, 52.00], \
    [2378, 'Reibel', 'Lothar', '', '97905078', 75000092, 52.00], \
    [2379, 'Gädecke', 'Burghard', '02262-93663', '90655625', 18750023, 34.00], \
    [2380, 'Groß', 'Hans', '', '49092536', 37500046, 17.00], \
    [2381, 'Kreuziger', 'Uwe', '02205-3956', '98258423', 75000092, 34.00], \
    [2382, 'Wegener', 'Ernst', '05733-3146', '29326533', 75000092, 52.00], \
    [2383, 'Schmanck', 'Heinrich', '02307-148533', '91041377', 56250069, 52.00], \
    [2384, 'Ferdinand', 'Karl', '02235-420400', '32434917', 56250069, 17.00], \
    [2385, 'Fichter', 'Manfred', '0201-210007', '33650415', 18750023, 52.00], \
    [2386, 'Sturm', 'Juergen', '02451-43040', '62537996', 37500046, 17.00], \
    [2387, 'Kaiser', 'Gerhard', '06252-6101', '76106393', 75000092, 34.00], \
    [2388, 'Lünskens', 'Klaus', '02041-703304', '20158893', 75000092, 52.00], \
    [2389, 'Wielens', 'Hans', '', '10518884', 93750115, 17.00], \
    [2390, 'Czischke', 'Heinrich', '', '90885151', 37500046, 52.00], \
    [2391, 'Böttger', 'Ulrich', '', '6738901', 37500046, 52.00], \
    [2392, 'Hunger', 'Adalbert', '02307-982800', '60724006', 37500046, 52.00], \
    [2393, 'Geißler', 'Nikolaus', '06251-4036', '88376157', 56250069, 17.00], \
    [2394, 'Stratmann', 'Werner', '02323-31651', '38071933', 75000092, 17.00], \
    [2395, 'Verspohl', 'Werner', '02841-23947', '74994153', 75000092, 52.00], \
    [2396, 'Horstmann', 'Manfred', '02307-6369', '26740520', 18750023, 34.00], \
    [2397, 'Windisch-Spoerk', 'Reinhold', '0043-1-8942141', '27935196', 37500046, 34.00], \
    [2398, 'Rinnen', 'Hans-Peter', '', '6850240', 18750023, 17.00], \
    [2399, 'Sombetzki', 'Volker', '', '31992460', 18750023, 17.00], \
    [2400, 'Schaak', 'Lothar', '069-851081', '97291833', 75000092, 52.00], \
    [2401, 'Dege', 'Werner', '02307-1480', '83837589', 93750115, 34.00], \
    [2402, 'Reeske', 'Werner', '', '58371773', 75000092, 34.00], \
    [2403, 'Wiegand', 'Werner', '05661-3550', '54394572', 56250069, 17.00], \
    [2404, 'Rockstroh', 'Werner', '069-21236964', '9571951', 37500046, 52.00], \
    [2405, 'Gerstner', 'Friedhelm', '02871-953250', '75793430', 18750023, 34.00], \
    [2406, 'Lohmann', 'Gerald', '02246-7868', '60179770', 18750023, 17.00], \
    [2407, 'Rode', 'Hanna', '0231-143041', '41404914', 93750115, 34.00], \
    [2408, 'Kordel', 'Rolf', '05662-3011', '92401960', 18750023, 17.00], \
    [2409, 'Schulze', 'Hein', '', '99745236', 37500046, 34.00], \
    [2410, 'Bender', 'Werner', '02304-16010', '18181638', 75000092, 52.00], \
    [2411, 'Mänken', 'Manfred', '02303-103490', '78800191', 37500046, 52.00], \
    [2412, 'Brune', 'Richard', '05681-99480', '94819114', 93750115, 52.00], \
    [2413, 'Elfein', 'Gerhard', '02521-9120', '24849483', 75000092, 52.00], \
    [2414, 'Keim', 'Günter', '07159-924760', '45260946', 37500046, 34.00], \
    [2415, 'Stier', 'Johannes', '02331-87603', '721122', 56250069, 17.00], \
    [2416, 'Katschke', 'Manfred', '02773-3054', '19902130', 18750023, 17.00], \
    [2417, 'Lasalle', 'Werner', '030-3309060', '95104267', 56250069, 34.00], \
    [2418, 'Köppen', 'Hans', '0251-42023', '17422180', 18750023, 17.00], \
    [2419, 'Schubert', 'Michael', '0231-5025930', '14591429', 93750115, 17.00], \
    [2420, 'Pagel', 'Werner', '02597-9579', '38152181', 37500046, 17.00], \
    [2421, 'Klatthaar', 'Rudi', '02306-6066', '23717063', 75000092, 34.00], \
    [2422, 'Gerstmeier', 'Josef', '02303-50381', '12693905', 18750023, 34.00], \
    [2423, 'Schmitz', 'Hermann', '02306-3175', '46472260', 18750023, 17.00], \
    [2424, 'Döll', 'Werner', '602206', '59274687', 56250069, 17.00], \
    [2425, 'Reindl', 'Jürgen', '09561-3170', '29676042', 56250069, 17.00], \
    [2426, 'Beckmann', 'Peter', '0911-5880262', '65247264', 93750115, 34.00], \
    [2427, 'Baumgärtner', 'Hannelore', '02045-7138', '10776601', 37500046, 52.00], \
    [2428, 'Blum', 'Hans-Dieter', '02661-6820', '22348520', 18750023, 34.00], \
    [2429, 'Seth', 'Helmut', '02058-3315', '11590238', 75000092, 34.00], \
    [2430, 'Hütte', 'Anneliese', '05031-3481', '97145172', 56250069, 17.00], \
    [2431, 'Bertram', 'Werner', '02464-7000', '18908504', 93750115, 17.00], \
    [2432, 'Bolz', 'Werner', '0221-5991431', '95228064', 93750115, 34.00], \
    [2433, 'Stelzmann', 'Gerold', '05653-97880', '98079666', 37500046, 34.00], \
    [2434, 'Bittner', 'Günter', '02951-2023', '24952423', 75000092, 17.00], \
    [2435, 'Freytag', 'Heinz', '02371-3854', '75389440', 18750023, 17.00], \
    [2436, 'Schmidt', 'Harry', '02303-55535415', '70319203', 75000092, 17.00], \
    [2437, 'Müller', 'Christian', '04221-66152', '54847853', 75000092, 34.00], \
    [2438, 'Meier', 'Ottmar', '02331-23820', '27867792', 56250069, 17.00], \
    [2439, 'Pieper', 'Hanneli', '05682-3866', '73993464', 93750115, 34.00], \
    [2440, 'Trummer', 'Wilhelm', '0211-279685', '8157862', 56250069, 34.00], \
    [2441, 'Scholl', 'Werner', '09664-510', '9191020', 18750023, 17.00], \
    [2442, 'Frauenrath', 'Richard', '', '73486842', 56250069, 17.00], \
    [2443, 'Stamen', 'Adalbert', '02402-750103', '93353770', 18750023, 17.00], \
    [2444, 'Rietschel', 'Hans-Georg', '0231-423050', '96820523', 75000092, 34.00], \
    [2445, 'Landfried', 'Bernd', '', '62532905', 18750023, 34.00], \
    [2446, 'van der Kolk', 'Werner', '07321-49905', '90550872', 56250069, 17.00], \
    [2447, 'Goldowsky', 'Ernst', '02263-3178', '88520248', 75000092, 17.00], \
    [2448, 'Laurenz', 'Wolfgang', '0345-2021255', '59818244', 93750115, 17.00], \
    [2449, 'Althoff', 'Friedhelm', '0251-784140', '65238003', 75000092, 52.00], \
    [2450, 'Bretz', 'Werner', '02303-103439', '55129683', 75000092, 52.00], \
    [2451, 'Helms', 'Meinhard', '06134-603410', '64586823', 75000092, 52.00], \
    [2452, 'Haß', 'Werner', '02594-4893', '10745085', 18750023, 52.00], \
    [2453, 'Olsen', 'Theo', '0234-233033', '92646544', 93750115, 52.00], \
    [2454, 'Trenkmann', 'Werner', '0212-2902421', '10474469', 93750115, 17.00], \
    [2455, 'Habermehl', 'Carsten', '03949-2347', '44409964', 93750115, 52.00], \
    [2456, 'Autenrieth', 'Carsten', '02541-3700', '98274700', 18750023, 17.00], \
    [2457, 'Trappe', 'Kuno', '0211-4350172', '2895705', 18750023, 52.00], \
    [2458, 'Roller', 'Jürgen', '03841-763104', '68118364', 93750115, 52.00], \
    [2459, 'Roller', 'Veronika', '', '65351179', 93750115, 52.00], \
    [2460, 'Weber', 'Werner', '', '57281773', 75000092, 17.00], \
    [2461, 'Bachem', 'Werner', '02961-794210', '34768291', 37500046, 52.00], \
    [2462, 'Duda', 'Norbert', '0221-634031', '50672654', 93750115, 17.00], \
    [2464, 'Buchholz', 'Paul', '03731-23092', '90924072', 56250069, 17.00], \
    [2465, 'Wrede', 'Werner', '0221-82752305', '60891888', 75000092, 52.00], \
    [2466, 'Dutz', 'Carsten', '', '9977934', 93750115, 34.00], \
    [2467, 'Schlinkert', 'Herbert', '02402-23791', '99142248', 75000092, 52.00], \
    [2468, 'Wommer', 'Werner', '', '83778518', 75000092, 34.00], \
    [2469, 'Roller', 'Werner', '02359-661202', '65552222', 56250069, 52.00], \
    [2470, 'Horak', 'Hans', '', '36203218', 75000092, 17.00], \
    [2471, 'Schmaußer', 'Joachim', '2268127', '67501175', 18750023, 34.00], \
    [2472, 'Ludwig', 'Werner', '06503-7518', '92253188', 75000092, 34.00], \
    [2473, 'Schulte', 'Rolf', '', '99340783', 75000092, 17.00], \
    [2474, 'Klein', 'Winfried', '089-12652738', '20551834', 93750115, 52.00], \
    [2475, 'Hille', 'Gisela', '05692-361', '14680840', 18750023, 17.00], \
    [2478, 'Zinnhardt', 'Rainer', '02330-611293', '57200590', 18750023, 17.00], \
    [2479, 'Alt', 'Werner', '02302-60051', '70327683', 75000092, 52.00], \
    [2480, 'Hofrath', 'Heinz', '06236-691162', '97379342', 56250069, 52.00], \
    [2481, 'Kraft', 'Werner', '0208-631454', '54604119', 93750115, 34.00], \
    [2482, 'Ludwig', 'Dieter', '0208-8500022', '31439683', 75000092, 17.00], \
    [2484, 'Strupp', 'Peter', '0381-7998860', '6288600', 18750023, 52.00], \
    [2485, 'Becker', 'Werner', '06131-572797', '92500493', 75000092, 34.00], \
    [2486, 'Ströhmann', 'Arno', '0621-2937647', '55790883', 75000092, 52.00], \
    [2487, 'Schweden', 'Werner', '02265-9672', '77384713', 75000092, 17.00], \
    [2488, 'Stiebing', 'Wolfgang', '02064-92085', '4887080', 18750023, 34.00], \
    [2489, 'Heinrichs', 'Rudolf', '069-80652635', '37163878', 75000092, 17.00], \
    [2490, 'Sandau', 'Manfred', '02264-45860', '13304870', 18750023, 34.00], \
    [2491, 'Stöcker', 'Hermann', '02261-42425', '14961434', 93750115, 17.00], \
    [2492, 'Dieterle', 'Frau', '0231-5023154', '56831798', 75000092, 34.00], \
    [2493, 'Brustat', 'Heino', '0711-35122478', '72034976', 37500046, 17.00], \
    [2494, 'Diester', 'Berthold', '02922-6004', '91845764', 93750115, 17.00], \
    [2495, 'Dix', 'Gerd', '0231-35831', '51756047', 56250069, 52.00], \
    [2496, 'Andres', 'Frau', '030-25887926', '63884544', 93750115, 34.00], \
    [2497, 'Feige', 'Werner', '07151-65102', '7532291', 37500046, 17.00], \
    [2498, 'Rittberger', 'Peter', '0034-43-212244', '95216817', 56250069, 17.00], \
    [2499, 'Renfordt', 'Gunther', '0711-95462210', '9051717', 56250069, 17.00], \
    [2500, 'Koch', 'Franz', '0541-16949', '32216738', 75000092, 34.00], \
    [2502, 'Tramski', 'Werner', '0041-31-8293110', '31277895', 18750023, 52.00], \
    [2503, 'Bäz', 'Peter', '09324-693', '87592808', 75000092, 34.00], \
    [2504, 'Hämmerle', 'Jo', '', '72328166', 37500046, 17.00], \
    [2505, 'Messner', 'Klaus P.', '0421-36116776', '19491420', 18750023, 52.00], \
    [2506, 'Kunze', 'Michael', '06132-99510', '92310756', 37500046, 34.00], \
    [2507, 'Friebe', 'Herr', '0611-317210', '66889770', 18750023, 52.00], \
    [2508, 'Wolff', 'Werner', '', '16379051', 37500046, 17.00], \
    [2509, 'Lohr', 'Claudia', '07274-1060', '82726781', 37500046, 17.00], \
    [2510, 'Welte', 'Eckhard', '02351-95930', '89010877', 56250069, 34.00], \
    [2511, 'Frankerl', 'Wolfgang', '07131-562452', '5817857', 56250069, 52.00], \
    [2512, 'Meyer', 'Gerlinde', '', '80736936', 37500046, 34.00], \
    [2513, 'Steinke', 'Werner', '05136-2433', '45687508', 75000092, 17.00], \
    [2514, 'Kurtz', 'Jürgen', '', '69958678', 75000092, 17.00], \
    [2515, 'Bund', 'Werner', '02301-14682', '95509201', 37500046, 52.00], \
    [2516, 'Post', 'Nadja', '07661-7971', '23147581', 37500046, 52.00], \
    [2517, 'Burg', 'Alois', '', '41398835', 18750023, 34.00], \
    [2518, 'Wernerus', 'Werner', '02372-10719', '90234970', 18750023, 17.00], \
    [2519, 'Henke', 'William', '02408-3071', '79151257', 56250069, 34.00], \
    [2520, 'Brinkkötter', 'Werner', '030-7848668', '95184640', 18750023, 17.00], \
    [2521, 'Baudis', 'Franz-Josef', '02596-4208', '68248036', 37500046, 52.00], \
    [2522, 'Schütz', 'Ekkehard', '', '44748488', 75000092, 34.00], \
    [2523, 'Störger', 'Werner', '', '87839318', 75000092, 34.00], \
    [2524, 'Holtkemper', 'Axel', '069-21233556', '15431958', 75000092, 52.00], \
    [2525, 'Nozulak', 'Werner', '02533-1266', '64934179', 93750115, 52.00], \
    [2526, 'Landgraf', 'Werner', '0561-63017', '48761058', 75000092, 52.00], \
    [2527, 'Rosen', 'Jens', '02304-40589', '27102076', 37500046, 52.00], \
    [2528, 'Küsters', 'Hans-Werner', '05971-50351', '4602212', 56250069, 52.00], \
    [2529, 'Strobel', 'Ingrid', '02451-8045', '49387179', 93750115, 34.00], \
    [2530, 'Scholze', 'Jens', '07940-149265', '50330961', 37500046, 34.00], \
    [2531, 'Holzgrefe', 'Hans', '07123-925243', '12471816', 37500046, 34.00], \
    [2532, 'Edeler', 'Heinz', '04321-9422427', '13817707', 56250069, 34.00], \
    [2533, 'Oehler', 'Otto', '0241-65046', '34208424', 93750115, 34.00], \
    [2534, 'Schneider', 'Michael', '02598-5345', '40985066', 37500046, 17.00], \
    [2535, 'Schwarz', 'Werner', '02432-2081', '36013604', 93750115, 17.00], \
    [2536, 'Röhnert', 'Jutta', '02371-69030', '80659180', 18750023, 17.00], \
    [2537, 'Siebert', 'Klara', '02382-59359', '38839587', 56250069, 17.00], \
    [2538, 'Mischke', 'Werner', '02403-24878', '76940753', 75000092, 34.00], \
    [2539, 'Weber', 'Frau', '04931-937214', '43248542', 56250069, 52.00], \
    [2540, 'Brinckwirth', 'Norbert', '06126-6064', '78253990', 18750023, 17.00], \
    [2541, 'Kespe', 'Friedrich', '02421-87741', '14819028', 75000092, 52.00], \
    [2542, 'Fuehr', 'Edwin', '0211-8999611', '63405987', 56250069, 17.00], \
    [2543, 'Rütten', 'Horst', '03606-67302', '64431300', 18750023, 52.00], \
    [2544, 'Schmidt', 'Werner', '0202-300477', '96875496', 37500046, 34.00], \
    [2545, 'Gipp', 'Werner', '0202-5636105', '86471798', 75000092, 34.00], \
    [2546, 'Flechsig', 'Karl-Heinz', '030-68092422', '31514999', 93750115, 17.00], \
    [2547, 'Steinhoff', 'Michael', '02301-5702', '72589854', 93750115, 34.00], \
    [2548, 'Werner', 'Werner', '0251-211028', '17682414', 93750115, 34.00], \
    [2549, 'Wortmann', 'Werner', '', '88343802', 56250069, 17.00], \
    [2550, 'Weber', 'Jan-Henrik', '06503-3023', '94287971', 37500046, 17.00], \
    [2551, 'Finselbach', 'Richard', '0040 1 3 110060', '97936178', 75000092, 34.00], \
    [2552, 'Lotter', 'Friedrich', '06074-50949', '13349087', 56250069, 52.00], \
    [2553, 'Preine', 'Karl-Josef', '', '58119023', 75000092, 34.00], \
    [2554, 'Schröder', 'Mike', '0511-2123930  F', '6427605', 18750023, 52.00], \
    [2555, 'Büchner', 'Werner', '0271-43488', '28767198', 75000092, 52.00], \
    [2556, 'Löffelholz', 'Egon', '', '90312634', 93750115, 52.00], \
    [2557, 'Wohne', 'Walter', '', '84444862', 56250069, 34.00], \
    [2558, 'Höppner', 'Doris', '', '64797839', 93750115, 17.00], \
    [2559, 'Hoch', 'Bärbel', '', '57291707', 56250069, 52.00], \
    [2560, 'Hinz', 'Fettin', '', '1725878', 75000092, 34.00], \
    [2561, 'Lichnock', 'Werner', '', '6845776', 37500046, 52.00], \
    [2562, 'Eccarius', 'Peter', '', '56966537', 56250069, 52.00], \
    [2563, 'Ute', 'Hartmut', '', '36593356', 37500046, 52.00], \
    [2564, 'Rademacher', 'Paul', '', '24537431', 37500046, 17.00], \
    [2565, 'Enders', 'Hans', '', '612115', 18750023, 17.00], \
    [2566, 'Stein', 'Hans', '', '87826567', 56250069, 34.00], \
    [2567, 'Witzel', 'Frau', '', '1124054', 93750115, 17.00], \
    [2568, 'Kilian', 'Ella', '', '91989013', 75000092, 17.00], \
    [2569, 'Zänker', 'Karl-Heinz', '', '59344455', 18750023, 52.00], \
    [2570, 'Körner', 'Frau', '', '2269643', 75000092, 34.00], \
    [2571, 'Bocke', 'Margrit', '', '45893602', 56250069, 34.00], \
    [2572, 'Löffler', 'Christa', '', '33779794', 93750115, 52.00], \
    [2573, 'Leichner', 'Werner', '', '37014083', 75000092, 34.00], \
    [2574, 'Eismann', 'Werner', '', '95702754', 93750115, 34.00], \
    [2575, 'Weiß', 'Werner', '', '76195964', 93750115, 17.00], \
    [2576, 'Schmidt', 'Werner', '', '25577455', 18750023, 17.00], \
    [2577, 'Müller', 'Uwe', '', '80297577', 56250069, 17.00], \
    [2578, 'Hirsch', 'Werner', '', '61380542', 56250069, 52.00], \
    [2579, 'Anlauf', 'Frau', '', '22779654', 93750115, 34.00], \
    [2580, 'Kühne', 'Werner', '', '57651435', 18750023, 52.00], \
    [2581, 'Meisgeier', 'Werner', '03378-870009', '47680656', 37500046, 34.00], \
    [2582, 'Schütze', 'Werner', '', '68794484', 93750115, 17.00], \
    [2584, 'Mense', 'Werner', '', '16925645', 18750023, 34.00], \
    [2585, 'Dr. Gebert', 'Gaby', '', '69324496', 37500046, 52.00], \
    [2586, 'Donszick', 'Werner', '', '36809777', 56250069, 52.00], \
    [2587, 'Borufka', 'Frauke', '', '61934113', 75000092, 17.00], \
    [2588, 'Rackwitz', 'Georg', '', '26643773', 75000092, 34.00], \
    [2589, 'Meisgeier', 'Werner', '', '1898979', 93750115, 34.00], \
    [2595, 'Liesch', 'Werner', '', '86160308', 75000092, 34.00], \
    [2596, 'Jessen-Lenz', 'Werner', '', '96309465', 18750023, 52.00], \
    [2597, 'Eckhardt', 'Werner', '', '73190473', 75000092, 17.00], \
    [2598, 'Meisgeier', 'Stefan', '', '12286099', 93750115, 52.00], \
    [2599, 'Kosanke', 'Werner', '', '92477727', 56250069, 17.00], \
    [2601, 'Wolf', 'Werner', '', '81768615', 18750023, 52.00], \
    [2602, 'Lamm', 'Werner', '', '85124433', 75000092, 52.00], \
    [2604, 'Lorenz', 'Werner', '', '69791065', 18750023, 17.00], \
    [2605, 'Müller', 'Werner', '', '78445935', 18750023, 52.00], \
    [2611, 'Hohne', 'Werner', '', '89160251', 37500046, 17.00], \
    [2612, 'Loibl', 'Ulrich', '03586-387108', '16348849', 93750115, 52.00], \
    [2613, 'Holzmüller', 'Anton', '', '75110818', 75000092, 17.00], \
    [2617, 'Kuhnert', 'Werner', '', '22499126', 37500046, 17.00], \
    [2624, 'Borsch', 'Werner', '', '5117224', 93750115, 52.00], \
    [2632, 'Freyer', 'Werner', '', '37129429', 93750115, 52.00], \
    [2634, 'Kühne', 'Karl', '', '78239073', 75000092, 52.00], \
    [2636, 'Lubitz', 'Carsten', '03445-703039', '81900609', 93750115, 34.00], \
    [2637, 'Effertz', 'Carsten', '03744-212527', '3540105', 18750023, 52.00], \
    [2638, 'Kistner', 'Carsten', '0221-2214228', '98549460', 18750023, 52.00], \
    [2639, 'Roller', 'Carsten', '05694-377', '20788536', 37500046, 34.00], \
    [2640, 'Roller', 'Ulrich', '', '92940165', 18750023, 52.00], \
    [2641, 'Roller', 'Herr', '', '45354039', 93750115, 34.00], \
    [2642, 'Roller', 'Fritz', '', '65342254', 93750115, 52.00], \
    [2643, 'Arndt', 'Ingrid', '', '7279232', 56250069, 52.00], \
    [2644, 'Gehrmann', 'Frauke', '02472-3061', '42059402', 56250069, 52.00], \
    [2645, 'Henneböhl', 'Werner', '06241-44638', '92385868', 75000092, 17.00], \
    [2646, 'Reinbender', 'Werner', '02904-808174', '29286262', 56250069, 34.00], \
    [2647, 'Dix', 'Herr', '04532-7611', '14221659', 93750115, 34.00], \
    [2648, 'Nowak-Fischer', 'Erika', '030-25887926', '23438968', 75000092, 17.00], \
    [2649, 'von der Heyden', 'Ludger', '06257-930214', '38750182', 56250069, 34.00], \
    [2650, 'Dragendorf', 'Bernd', '0271-392480', '74319335', 18750023, 34.00], \
    [2651, 'Grupp', 'Carsten', '', '85075331', 37500046, 17.00], \
    [2652, 'Quante', 'Werner', '07041-83343', '61100751', 37500046, 34.00], \
    [2653, 'Scholl', 'Heiner', '', '20335425', 18750023, 52.00], \
    [2655, 'Roller', 'Joachim', '', '61139995', 18750023, 17.00], \
    [2657, 'Ziegler-Möller', 'Friedrich-Wilhelm', '', '58816080', 18750023, 52.00], \
    [2658, 'Hoffecker', 'Helge', '05651-10851', '21865320', 18750023, 52.00], \
    [2659, 'Bliemeister', 'Arkadius', '', '90441563', 75000092, 34.00], \
    [2660, 'Weißbach', 'Dieter', '06251-51656', '51365372', 56250069, 52.00], \
    [2661, 'Habermehl', 'Helmut', '', '97813527', 56250069, 17.00], \
    [2662, 'Mientus', 'Hermann', '', '51950731', 37500046, 52.00], \
    [2663, 'Köhler', 'Werner', '05452-2552', '35947091', 37500046, 17.00], \
    [2664, 'Trowe', 'Thomas', '02330-71572', '62767204', 93750115, 52.00], \
    [2666, 'Diester', 'Heinz', '', '81776510', 18750023, 34.00], \
    [2667, 'Schobbe', 'Karl-Hermann', '02306-50519', '17815045', 18750023, 17.00], \
    [2668, 'Nick', 'Erich', '06103-62071', '17470643', 75000092, 34.00], \
    [2669, 'Brockhoff', 'Werner', '04331-5071', '29105157', 56250069, 17.00], \
    [2670, 'Rollke', 'Werner', '02371-44466', '70377605', 18750023, 34.00], \
    [2671, 'Jännert', 'Helmut', '02933-4015', '36413867', 56250069, 52.00], \
    [2672, 'Rösch', 'Klaus', '02845-32693', '7873574', 93750115, 17.00], \
    [2673, 'Göbel', 'Jörg', '', '31171456', 37500046, 52.00], \
    [2674, 'Nübel', 'Josef', '', '98673411', 37500046, 34.00], \
    [2675, 'Sennholz', 'Wolfgang', '02331-63932', '4019332', 56250069, 34.00], \
    [2676, 'Reckmann', 'Werner', '0231-254307', '52925901', 37500046, 34.00], \
    [2677, 'Post', 'Carsten', '02303-40043', '25621235', 18750023, 34.00], \
    [2678, 'Hinners', 'Hartmut', '02373-2243', '16107770', 18750023, 34.00], \
    [2679, 'Bachmann', 'Elke', '040-6039276', '23300841', 37500046, 34.00], \
    [2680, 'Roller', 'Jürgen', '0361-6463129', '17235400', 18750023, 17.00], \
    [2681, 'Fastenrath', 'Werner', '', '75374456', 37500046, 17.00], \
    [2682, 'Karasek', 'Thomas', '07473-1070', '18486988', 75000092, 17.00], \
    [2683, 'Schomaker', 'Werner', '', '10823655', 18750023, 52.00], \
    [2684, 'Feldmüller', 'Joachim', '0231-285986', '13173394', 93750115, 52.00], \
    [2685, 'Bös', 'Werner', '02306-54056', '5238510', 18750023, 52.00], \
    [2686, 'Rehbein', 'Werner', '06621-76282', '43652755', 18750023, 17.00], \
    [2687, 'Fehrmann', 'Werner', '040-4604659', '71868867', 56250069, 17.00], \
    [2688, 'Bänsch', 'Michael', '', '53062', 56250069, 34.00], \
    [2690, 'Becker', 'Werner', '', '4651356', 37500046, 34.00], \
    [2691, 'Becker', 'Hartmut', '', '75343059', 93750115, 34.00], \
    [2692, 'Haensch', 'Bernd', '0621-2937647', '90802650', 18750023, 52.00], \
    [2693, 'Henning', 'Peter', '', '20149669', 93750115, 52.00], \
    [2694, 'Türk', 'Klaus-Jürgen', '', '55490248', 75000092, 17.00], \
    [2695, 'Laser', 'Friedhelm', '0231-5024450', '24608717', 56250069, 52.00], \
    [2696, 'Pottmeyer', 'Werner', '0231-5024360', '98714201', 37500046, 17.00], \
    [2697, 'Hübner', 'Walter', '0231-5027014', '89585800', 18750023, 17.00], \
    [2698, 'Siebert', 'Walter', '05042-2055', '68553878', 75000092, 34.00], \
    [2699, 'Grote', 'Veronika', '05251-35093', '85029553', 75000092, 17.00], \
    [2700, 'Aluttis', 'Thomas', '02922-5017', '91713869', 93750115, 17.00], \
    [2701, 'Schnitzspan', 'Werner', '0231-423040', '34996434', 93750115, 34.00], \
    [2702, 'Duda', 'Gabriele', '06150-13050', '85080182', 56250069, 52.00], \
    [2703, 'Kuhn', 'Werner', '0261-31336', '42620996', 37500046, 17.00], \
    [2704, 'Kreuziger', 'Franz', '07903-644', '31090503', 75000092, 52.00], \
    [2705, 'Schneider', 'Rotraud', '02205-3956', '38085724', 93750115, 52.00], \
    [2706, 'Sandau', 'Werner', '', '6779123', 75000092, 34.00], \
    [2707, 'Wahle', 'Klaus', '02261-42425', '36720390', 18750023, 52.00], \
    [2708, 'Grün-Wenkel', 'Bernd', '0291-99680', '59596792', 56250069, 34.00], \
    [2709, 'von der Heyden', 'Werner', '069-21230855', '43292272', 56250069, 34.00], \
    [2710, 'Mühlenbrock', 'Wolfgang', '0271-392480', '13218936', 37500046, 34.00], \
    [2711, 'Steffen', 'Gerda', '0203-2834936', '70658207', 56250069, 52.00], \
    [2712, 'Feldmüller', 'Eckhard', '', '52486936', 37500046, 52.00], \
    [2713, 'Kuhn', 'Darko', '02306-54056', '47940073', 75000092, 17.00], \
    [2714, 'Reinecke', 'Hartmut', '0561-18265', '28820188', 75000092, 17.00], \
    [2715, 'Steinke', 'Werner', '02739-2856', '80908845', 18750023, 52.00], \
    [2716, 'Heimbring', 'Hartmut', '02309-74348', '34970162', 56250069, 52.00], \
    [2717, 'Fastenrath', 'Judith', '02403-23049', '43513297', 56250069, 34.00], \
    [2718, 'Jung-Schinauer', 'Werner', '07473-1070', '27998843', 75000092, 34.00], \
    [2719, 'Spenn', 'Erna', '02762-8755', '76570130', 18750023, 34.00], \
    [2720, 'Ullrich-Borrmann', 'Helene', '0214-52015', '14182551', 37500046, 34.00], \
    [2721, 'Klein', 'Klaus', '069-21233840', '66137433', 75000092, 52.00], \
    [2722, 'Prange', 'Gunter', '0211-8928750', '71844084', 93750115, 34.00], \
    [2723, 'Keller', 'Peter', '05681-7073', '83532124', 93750115, 52.00], \
    [2724, 'Scheuermann', 'Helmut', '069-21235334', '24420674', 93750115, 17.00], \
    [2725, 'Eckart', 'Rainer', '0361-7921412', '38283072', 56250069, 17.00], \
    [2726, 'Becker', 'Hans', '06061-2425', '23592067', 56250069, 34.00], \
    [2727, 'Damm', 'Carsten', '0231-5023136', '11874127', 56250069, 34.00], \
    [2728, 'Katriniok', 'Carsten', '069-471303', '68461450', 18750023, 17.00], \
    [2729, 'Geißler', 'Carsten', '02041-25289', '31606390', 18750023, 17.00], \
    [2730, 'Roller', 'Michael', '06251-4036', '8604042', 56250069, 17.00], \
    [2731, 'Roller', 'Walter', '', '8136295', 18750023, 17.00], \
    [2732, 'Roller', 'Rolf', '', '46725250', 18750023, 17.00], \
    [2733, 'Weber', 'Werner', '', '11289452', 56250069, 52.00], \
    [2734, 'Aluttis', 'Dieter', '0040 1 3 110060', '41824823', 75000092, 34.00], \
    [2735, 'Zinnhardt', 'Josef', '0231-423040', '49800699', 93750115, 34.00], \
    [2736, 'Staps', 'Heinz', '02302-60051', '359741', 37500046, 17.00], \
    [2737, 'Wulf', 'Fritz', '03441-213082', '46170785', 18750023, 34.00], \
    [2738, 'Hess', 'Udo', '02303-271244', '92391660', 18750023, 52.00], \
    [2739, 'Strupp', 'Bernd', '040-810181', '26640035', 18750023, 34.00], \
    [2740, 'Krumm', 'Heinz', '06131-572797', '39075930', 18750023, 52.00], \
    [2741, 'Wehr', 'Ludger', '02191-667811', '49162817', 56250069, 52.00], \
    [2742, 'Riechert', 'Arne', '02191-667811', '79471204', 93750115, 52.00], \
    [2743, 'Meier', 'Karl-Hermann', '', '72538953', 75000092, 52.00], \
    [2744, 'Quante', 'Werner', '05682-3866', '62185904', 93750115, 17.00], \
    [2745, 'Schroeter', 'Dieter', '', '58490767', 56250069, 34.00], \
    [2746, 'Rollke', 'Werner', '', '54321493', 75000092, 17.00], \
    [2747, 'Krekeler', 'Hans', '02933-4015', '81340692', 56250069, 17.00], \
    [2748, 'Vahrenhold', 'Reinhard', '2304', '93011182', 56250069, 34.00], \
    [2749, 'Friede', 'Werner', '02363-37470', '38941093', 75000092, 17.00], \
    [2750, 'Schneider', 'Werner', '0203-781531', '82951508', 75000092, 34.00], \
    [2751, 'Sell', 'Hans', '02432-2081', '65998681', 37500046, 52.00], \
    [2752, 'Neßler', 'Hans', '040-6905032', '39838837', 56250069, 34.00], \
    [2753, 'Smolka', 'Roland', '069-21233525', '71336018', 75000092, 34.00], \
    [2754, 'Schneider', 'Peter', '0211-8999611', '44458461', 37500046, 34.00], \
    [2755, 'Schneider', 'Dieter', '0241-474260', '13369145', 18750023, 34.00], \
    [2756, 'Stier', 'Helmut', '0241-474260', '95225009', 93750115, 17.00], \
    [2757, 'Becker', 'Richard', '02773-3054', '46730075', 18750023, 34.00], \
    [2758, 'Pferdekamp', 'Bernd', '0231-5023136', '92370637', 56250069, 34.00], \
    [2759, 'Damm', 'Fritz', '02590-1222', '73309986', 37500046, 34.00], \
    [2760, 'Grünbeck', 'Günter', '069-471303', '20938640', 18750023, 34.00], \
    [2761, 'Pauli', 'Anette', '09561-3170', '15726468', 75000092, 52.00], \
    [2762, 'Buckel', 'Friedhelm', '0211-428668', '14750375', 18750023, 34.00], \
    [2763, 'Clement', 'Burghard', '', '9562019', 93750115, 17.00], \
    [2764, 'Uckelmann', 'Werner', '06852-6272', '8864017', 56250069, 34.00], \
    [2765, 'Kordel', 'Thomas', '02591-6055', '77542686', 37500046, 34.00], \
    [2766, 'Schmanck', 'Michael', '', '59927284', 93750115, 52.00], \
    [2767, 'Haas', 'Meinolf', '02307-148533', '97873151', 37500046, 17.00], \
    [2768, 'Müller', 'Hans-Georg', '06131-71157', '54061748', 75000092, 34.00], \
    [2769, 'Dietrich', 'Wolfgang', '02922-861562', '15800528', 75000092, 34.00], \
    [2770, 'Dietrich', 'Horst', '03735-91520', '59476491', 37500046, 34.00], \
    [2771, 'Thöne', 'Jo', '03735-91520', '21386557', 56250069, 34.00], \
    [2772, 'Helms', 'Werner', '02841-94270', '43879753', 75000092, 17.00], \
    [2773, 'Flechsig', 'Karl', '02594-4893', '87014134', 93750115, 52.00], \
    [2774, 'Weber', 'Gert', '02301-5702', '46560133', 75000092, 17.00], \
    [2775, 'Deike', 'Darko', '06126-6064', '36160655', 18750023, 34.00], \
    [2776, 'Pfeiffer', 'Werner', '0208-4554840', '52140502', 56250069, 34.00], \
    [2777, 'Strobel', 'Hermann', '07954-98020', '25419529', 93750115, 52.00], \
    [2778, 'Heimbring', 'Ottmar', '', '26031401', 37500046, 17.00], \
    [2779, 'Göbel', 'Carsten', '02403-23049', '9758516', 37500046, 17.00], \
    [2780, 'Daldrup', 'Rudolf', '', '11389142', 56250069, 52.00], \
    [2781, 'Scholl', 'Werner', '0231-9172190', '17310389', 93750115, 17.00], \
    [2782, 'Roller', 'Hans-Werner', '', '35715743', 75000092, 34.00], \
    [2783, 'Bubeck', 'Andreas', '', '74506430', 18750023, 34.00], \
    [2784, 'Scheiber', 'Andreas', '07156-205280', '29578234', 93750115, 52.00], \
    [2785, 'Holzgrefe', 'Werner', '', '21388056', 37500046, 34.00], \
    [2786, 'Putz', 'Werner', '04321-9422427', '69063444', 93750115, 34.00], \
    [2787, 'Putz', 'Wilhelm', '0421-3694180', '49308821', 37500046, 17.00], \
    [2788, 'Küsters', 'Werner', '0421-3694180', '33365892', 56250069, 17.00], \
    [2789, 'Lauterjung', 'Richard', '02451-8045', '80555158', 75000092, 17.00], \
    [2790, 'Stamen', 'Werner', '', '90557047', 56250069, 34.00], \
    [2791, 'Heise', 'Siegfried', '0231-423050', '92289295', 18750023, 17.00], \
    [2792, 'Schröder', 'Rainer', '06063-912325', '6197456', 37500046, 17.00], \
    [2793, 'Metzner', 'Peter', '0271-43488', '60644455', 18750023, 17.00], \
    [2794, 'Buss', 'Andrea', '0351-8485323', '51363843', 75000092, 52.00], \
    [2795, 'Schneider', 'Sabine', '02374-3065', '72110947', 56250069, 34.00], \
    [2796, 'Kunze', 'Dieter', '02389-71550', '68700690', 18750023, 52.00], \
    [2797, 'Strelow', 'Günter', '0611-317210', '52508470', 18750023, 17.00], \
    [2798, 'Leifhelm', 'Heinz', '', '9868954', 93750115, 52.00], \
    [2799, 'Wulf', 'Michael', '02862-42150', '67525311', 37500046, 34.00], \
    [2800, 'Strobel', 'Manfred', '02303-271244', '97912690', 18750023, 17.00], \
    [2801, 'Krost', 'Gudula', '07940-149265', '18590935', 18750023, 17.00], \
    [2802, 'Fritze', 'Carsten', '0177-7154189', '57853976', 37500046, 17.00], \
    [2803, 'Schubert', 'Heinz', '05426-4545', '51778749', 93750115, 34.00], \
    [2804, 'Farwig', 'Heinz', '02597-9579', '45438046', 37500046, 52.00], \
    [2805, 'Roller', 'Werner', '96196-3055', '77241183', 75000092, 52.00], \
    [2806, 'Janßen', 'Carsten', '', '92566066', 37500046, 52.00], \
    [2807, 'Janßen', 'Michael', '', '57878289', 93750115, 34.00], \
    [2808, 'Messner', 'Werner', '', '60405227', 56250069, 52.00], \
    [2809, 'Roller', 'Bernd', '06132-99510', '30663683', 75000092, 34.00], \
    [2810, 'Siebert', 'Hans-Joachim', '', '64520969', 93750115, 17.00], \
    [2811, 'Mischke', 'Albert', '02403-24878', '3928987', 56250069, 34.00], \
    [2812, 'Jähn', 'Albert', '04931-937214', '34494434', 93750115, 17.00], \
    [2813, 'Thomas', 'Giuseppina', '', '47962196', 37500046, 17.00], \
    [2814, 'Fehr', 'Werner', '02241-805422', '643165', 18750023, 17.00], \
    [2815, 'Fehr', 'Norbert', '0941-990551', '9685739', 93750115, 17.00], \
    [2816, 'Milazzo', 'Klaus', '0941-990551', '88222325', 18750023, 34.00], \
    [2817, 'Buchholz', 'Friedhelm', '07531-98250', '52630849', 93750115, 52.00], \
    [2818, 'Breuer', 'Heinz', '0221-82752305', '66681058', 75000092, 17.00], \
    [2819, 'Hirsch', 'Tammo', '02581-2260', '8553391', 37500046, 52.00], \
    [2820, 'Bott', 'Egon', '0621-504425620', '56854285', 18750023, 17.00], \
    [2821, 'Himmelreich', 'Doris', '', '49971130', 18750023, 17.00], \
    [2822, 'Ricklefs', 'Rotraud', '069-80652925', '64788976', 37500046, 52.00], \
    [2823, 'Aebersold', 'Ernst', '', '8325377', 56250069, 52.00], \
    [2824, 'Föhr', 'Werner', '0041-61-2959715', '24620140', 18750023, 17.00], \
    [2825, 'Störing', 'Wilhelm', '', '90703648', 75000092, 17.00], \
    [2826, 'Bretz', 'Manfred', '02552-3103', '56220519', 93750115, 34.00], \
    [2827, 'Jannek', 'Manfred', '06134-603410', '98273176', 37500046, 52.00], \
    [2828, 'Werner', 'Manfred', '06127-90700', '77280781', 37500046, 52.00], \
    [2829, 'Schubert', 'Werner', '06421-92840', '58091550', 18750023, 52.00], \
    [2830, 'Schubert', 'Bernd', '02597-9579', '44692167', 56250069, 17.00], \
    [2831, 'Schubert', 'Volker', '02593-6681', '41265001', 37500046, 52.00], \
    [2832, 'Wolf', 'Helmut', '02593-6681', '40876965', 18750023, 52.00], \
    [2833, 'Wießner', 'Werner', '', '19151721', 37500046, 34.00], \
    [2834, 'Jesinghausen', 'Thomas', '069-21235531', '82985338', 75000092, 17.00], \
    [2835, 'Bolz', 'Thomas', '06146-2078', '34592234', 93750115, 17.00], \
    [2836, 'Bachmann', 'Albert', '05653-97880', '91685226', 37500046, 34.00], \
    [2837, 'Pohl', 'Friedhelm', '0361-6463129', '53598809', 93750115, 17.00], \
    [2838, 'Pohl', 'Ulrich', '06074-7401', '35192832', 56250069, 17.00], \
    [2839, 'Lex', 'Hiltrud', '06074-7401', '85934376', 37500046, 34.00], \
    [2840, 'Haß', 'Hiltrud', '', '99070931', 37500046, 17.00], \
    [2841, 'Terhardt', 'Dr. Michael', '0234-233033', '85142673', 75000092, 52.00], \
    [2842, 'Westram', 'Werner', '02541-3706', '92915838', 75000092, 52.00], \
    [2843, 'Westram', 'Werner', '02235-952273', '36490421', 37500046, 17.00], \
    [2844, 'Just', 'Bernd', '02235-952273', '94521637', 56250069, 34.00], \
    [2845, 'Brönstrup', 'Ekkehard', '040-8891610', '25158183', 75000092, 52.00], \
    [2846, 'Krekeler', 'Joachim', '', '33597484', 93750115, 52.00], \
    [2847, 'Steffen', 'Werner', '2304', '72922429', 93750115, 52.00], \
    [2848, 'Nozulak', 'Werner', '0251-328174', '7265948', 75000092, 34.00], \
    [2849, 'Klein', 'Werner', '0561-63017', '18790306', 37500046, 52.00], \
    [2850, 'Krönung', 'Werner', '05692-361', '8673277', 56250069, 34.00], \
    [2851, 'Günther', 'Werner', '06438-2672', '71775747', 56250069, 17.00], \
    [2852, 'Wernerus', 'Werner', '0231-523444', '34988174', 93750115, 17.00], \
    [2853, 'Schmitz', 'Werner', '02408-3071', '46243522', 56250069, 34.00], \
    [2854, 'Königs', 'Hans', '602206', '14874452', 56250069, 52.00], \
    [2855, 'Giese', 'Norbert', '02506-81120', '94916892', 56250069, 17.00], \
    [2856, 'Steinhoff', 'Werner', '', '68205455', 18750023, 34.00], \
    [2857, 'Klatthaar', 'Dieter', '0251-211028', '29094326', 37500046, 17.00], \
    [2858, 'Dutz', 'Ulrich', '02303-50381', '88338604', 93750115, 52.00], \
    [2859, 'Adolphy', 'Ulrich', '02402-23791', '6922994', 93750115, 17.00], \
    [2860, 'Koch', 'Georg', '06732-2666', '54760914', 93750115, 34.00], \
    [2861, 'Halt', 'Jo', '06206-950724', '66697956', 37500046, 34.00], \
    [2862, 'Halt', 'Herr', '', '92561375', 18750023, 34.00], \
    [2863, 'Büscher', 'Franz-Josef', '', '79033097', 56250069, 52.00], \
    [2864, 'Hess', 'Giuseppina', '02521-9120', '9266137', 56250069, 34.00], \
    [2865, 'Gehrmann', 'Werner', '040-810181', '35966293', 75000092, 17.00], \
    [2866, 'Holtkemper', 'Fritz', '06241-44638', '85496932', 56250069, 34.00], \
    [2867, 'Milazzo', 'Rotraud', '02533-1266', '46949126', 37500046, 17.00], \
    [2868, 'Szonneck', 'Michael', '07531-98250', '73963609', 93750115, 52.00], \
    [2869, 'Henneböhl', 'Rainer', '02371-438710', '6385067', 56250069, 52.00], \
    [2870, 'Grün-Wenkel', 'Winfried', '02904-808174', '13826019', 93750115, 34.00], \
    [2871, 'Lohr', 'Johann', '069-21230855', '73904419', 93750115, 52.00], \
    [2872, 'Kraft', 'Jakob', '02351-95930', '11580749', 93750115, 17.00], \
    [2873, 'Alt', 'Heidrun', '0208-8500022', '37651593', 75000092, 52.00], \
    [2874, 'Eckardt', 'Hans', '06236-691162', '10323934', 93750115, 52.00], \
    [2875, 'Eckardt', 'Werner', '0231-992090', '14367051', 37500046, 34.00], \
    [2876, 'Sarges', 'Klemens', '0231-992090', '67041198', 75000092, 52.00], \
    [2877, 'Göpfrich', 'Klaus', '0641-3062569', '61421201', 37500046, 17.00], \
    [2878, 'Hartmann', 'Wolf', '07161-962950', '6131259', 93750115, 34.00], \
    [2879, 'Remfort', 'Willi', '', '48085368', 75000092, 52.00], \
    [2880, 'Brinckwirth', 'Peter', '05261-25840', '70782793', 75000092, 17.00], \
    [2881, 'Lehnen', 'Veronika', '02421-87741', '46413248', 75000092, 34.00], \
    [2882, 'Lehnen', 'Wolfgang', '', '67467086', 37500046, 17.00], \
    [2883, 'Borbonus', 'Christine', '', '19320696', 37500046, 34.00], \
    [2884, 'Völker', 'Ingrid', '', '22944647', 56250069, 52.00], \
    [2885, 'Kurtz', 'Hans-Jürgen', '06142-301730', '30698418', 75000092, 52.00], \
    [2886, 'Brink', 'Ulrike', '02301-14682', '48313331', 37500046, 17.00], \
    [2887, 'Edeler', 'Werner', '', '2159950', 18750023, 17.00], \
    [2890, 'Smula', 'Franz', '06174-20000', '7183451', 37500046, 17.00], \
    [2891, 'Renfordt', 'Franz', '0209-85809', '32271510', 18750023, 52.00], \
    [2892, 'Finselbach', 'Claus', '0541-16949', '87279613', 75000092, 17.00], \
    [2893, 'Hämmerle', 'Sabine', '06074-50949', '8456207', 56250069, 52.00], \
    [2894, 'Hämmerle', 'Karl', '0421-481202', '23481629', 93750115, 17.00], \
    [2895, 'Walther', 'Werner', '0421-481202', '55740305', 18750023, 34.00], \
    [2896, 'Neuser', 'Inge', '', '27320486', 37500046, 17.00], \
    [2897, 'Kistner', 'Werner', '', '75412795', 18750023, 17.00], \
    [2898, 'Sandkämper', 'Günter', '05694-377', '62851536', 37500046, 34.00], \
    [2899, 'Heidelberg', 'Werner', '', '45765952', 56250069, 34.00], \
    [2900, 'Rode', 'Norbert', '', '88363016', 37500046, 17.00], \
    [2902, 'Mänz', 'Ernst', '', '2226993', 75000092, 52.00], \
    [2903, 'Hager', 'Christian', '02401-2151', '29077224', 93750115, 34.00], \
    [2904, 'Breuer', 'Klaus', '', '917541', 37500046, 34.00], \
    [2905, 'Sturm', 'Werner', '02581-2260', '21114772', 56250069, 34.00], \
    [2906, 'Böttcher', 'Werner', '02451-43040', '75984899', 93750115, 17.00], \
    [2907, 'Brodde', 'Werner', '', '78198107', 56250069, 52.00], \
    [2908, 'Effertz', 'Werner', '', '75597584', 93750115, 17.00], \
    [2909, 'Wolf', 'Jörg', '0221-2214228', '68602954', 93750115, 52.00], \
    [2910, 'Heise', 'Werner', '0551-371667', '9849044', 93750115, 17.00], \
    [2911, 'Baumgärtner', 'Thomas', '', '32898719', 93750115, 17.00], \
    [2912, 'Dietrich', 'Werner', '02661-6820', '28781504', 93750115, 17.00], \
    [2913, 'Bänsch', 'Dieter', '069-21245478', '30753568', 75000092, 17.00], \
    [2914, 'Nick', 'Johannes', '02361-33166', '75218417', 56250069, 52.00], \
    [2915, 'Keller', 'Sigrid', '04331-5071', '95945783', 75000092, 34.00], \
    [2916, 'Endlich', 'Werner', '02594-3942', '60779646', 37500046, 34.00], \
    [2917, 'Calaminus', 'Otto', '06131-611012', '23516931', 37500046, 34.00], \
    [2918, 'Grünig', 'Theo', '040-407973', '49072229', 93750115, 17.00], \
    [2919, 'Freytag', 'Ulrich', '06131-687091', '56961417', 56250069, 17.00], \
    [2920, 'Röhnert', 'Ulrich', '02303-55535415', '50538512', 56250069, 52.00], \
    [2921, 'Autenrieth', 'Carsten', '02382-59359', '787903', 75000092, 17.00], \
    [2922, 'Formann', 'Nadja', '0211-4350172', '42291920', 18750023, 34.00], \
    [2923, 'Formann', 'Berthold', '', '301003', 75000092, 17.00], \
    [2924, 'Roller', 'Inge', '', '58965089', 93750115, 17.00], \
    [2925, 'Henke', 'Werner', '', '26255742', 56250069, 17.00], \
    [2926, 'Winterlich', 'Monika', '030-7848668', '80266869', 93750115, 34.00], \
    [2927, 'Klein', 'Werner', '0241-1769200', '45275317', 56250069, 34.00], \
    [2928, 'Schmidt', 'Jörn', '0201-230607', '42263226', 37500046, 34.00], \
    [2929, 'Buttgereit', 'Werner', '', '18042477', 56250069, 17.00], \
    [2930, 'Holzapfel', 'Werner', '', '39921043', 75000092, 17.00], \
    [2931, 'Brunken', 'Joachim', '02721-80104', '52983681', 37500046, 34.00], \
    [2932, 'Wommer', 'Johannes', '02721-80104', '96732786', 37500046, 34.00], \
    [2933, 'Stiebing', 'Ludwig', '02359-661202', '87936150', 18750023, 52.00], \
    [2934, 'Schirmeister', 'Wolfgang', '069-80652635', '42889445', 18750023, 34.00], \
    [2935, 'Köppen', 'Werner', '', '42904725', 18750023, 52.00], \
    [2936, 'Schöneberg', 'Heino', '0231-5025930', '99727033', 75000092, 17.00], \
    [2938, 'Stöcker', 'Werner', '', '892781', 37500046, 17.00], \
    [2939, 'Weishaupt', 'Hans-W', '0231-5023154', '4971849', 93750115, 34.00], \
    [2940, 'Andres', 'Werner', '', '61867529', 93750115, 17.00], \
    [2941, 'Schmidt', 'Ulrich', '07151-65102', '7566885', 18750023, 52.00], \
    [2942, 'Kramer', 'Anton', '', '76647964', 93750115, 52.00], \
    [2943, 'Schriek', 'Bernard', '04528-55478', '22331997', 56250069, 17.00]]

    cursor.executemany("INSERT INTO mitglied VALUES (?,?,?,?,?,?,?)",datalist)
    connection.commit()
    connection.close()

def teilnehmer_fuellen():
    connection = sq.connect("sportverein.db")
    cursor = connection.cursor()

    datalist = [[1963, 17],   \
    [1964, 17],   \
    [1965, 11],   \
    [1966, 17],   \
    [1967, 13],   \
    [1968, 1],   \
    [1969, 3],   \
    [1970, 13],   \
    [1971, 8],   \
    [1972, 3],   \
    [1973, 12],   \
    [1974, 3],   \
    [1975, 11],   \
    [1976, 4],   \
    [1977, 12],   \
    [1978, 2],   \
    [1979, 3],   \
    [1980, 16],   \
    [1981, 1],   \
    [1982, 4],   \
    [1983, 3],   \
    [1984, 13],   \
    [1985, 8],   \
    [1986, 1],   \
    [1988, 11],   \
    [1989, 12],   \
    [1990, 3],   \
    [1991, 3],   \
    [1992, 3],   \
    [1993, 14],   \
    [1994, 4],   \
    [1996, 17],   \
    [1997, 11],   \
    [1998, 16],   \
    [1999, 17],   \
    [2000, 7],   \
    [2001, 13],   \
    [2002, 11],   \
    [2004, 10],   \
    [2005, 13],   \
    [2007, 5],   \
    [2009, 2],   \
    [2010, 3],   \
    [2011, 12],   \
    [2012, 13],   \
    [2013, 5],   \
    [2014, 3],   \
    [2015, 6],   \
    [2016, 15],   \
    [2017, 4],   \
    [2018, 2],   \
    [2019, 4],   \
    [2020, 6],   \
    [2021, 2],   \
    [2022, 11],   \
    [2023, 7],   \
    [2024, 1],   \
    [2025, 8],   \
    [2026, 12],   \
    [2027, 6],   \
    [2028, 17],   \
    [2029, 6],   \
    [2030, 8],   \
    [2031, 7],   \
    [2032, 3],   \
    [2033, 14],   \
    [2034, 3],   \
    [2035, 12],   \
    [2036, 4],   \
    [2037, 14],   \
    [2038, 17],   \
    [2039, 4],   \
    [2040, 1],   \
    [2041, 12],   \
    [2042, 11],   \
    [2043, 8],   \
    [2044, 12],   \
    [2045, 6],   \
    [2046, 12],   \
    [2047, 8],   \
    [2048, 10],   \
    [2049, 2],   \
    [2050, 10],   \
    [2052, 3],   \
    [2053, 7],   \
    [2054, 7],   \
    [2055, 8],   \
    [2056, 16],   \
    [2057, 3],   \
    [2058, 12],   \
    [2059, 7],   \
    [2060, 14],   \
    [2061, 12],   \
    [2062, 14],   \
    [2063, 12],   \
    [2064, 2],   \
    [2065, 10],   \
    [2066, 6],   \
    [2067, 1],   \
    [2068, 17],   \
    [2069, 14],   \
    [2070, 4],   \
    [2071, 10],   \
    [2072, 1],   \
    [2073, 6],   \
    [2074, 3],   \
    [2075, 6],   \
    [2076, 12],   \
    [2077, 6],   \
    [2078, 6],   \
    [2079, 7],   \
    [2080, 12],   \
    [2081, 13],   \
    [2082, 9],   \
    [2083, 1],   \
    [2084, 10],   \
    [2085, 12],   \
    [2086, 5],   \
    [2087, 6],   \
    [2088, 7],   \
    [2089, 16],   \
    [2090, 8],   \
    [2091, 1],   \
    [2092, 6],   \
    [2093, 5],   \
    [2094, 13],   \
    [2095, 17],   \
    [2096, 3],   \
    [2098, 8],   \
    [2099, 5],   \
    [2101, 10],   \
    [2102, 1],   \
    [2104, 13],   \
    [2105, 3],   \
    [2106, 6],   \
    [2107, 3],   \
    [2108, 2],   \
    [2109, 4],   \
    [2110, 6],   \
    [2112, 15],   \
    [2113, 5],   \
    [2114, 5],   \
    [2115, 17],   \
    [2116, 11],   \
    [2117, 15],   \
    [2118, 11],   \
    [2119, 7],   \
    [2120, 14],   \
    [2121, 14],   \
    [2122, 13],   \
    [2123, 12],   \
    [2124, 17],   \
    [2125, 8],   \
    [2126, 8],   \
    [2128, 10],   \
    [2129, 3],   \
    [2130, 17],   \
    [2131, 7],   \
    [2132, 1],   \
    [2133, 9],   \
    [2134, 6],   \
    [2135, 4],   \
    [2136, 4],   \
    [2137, 4],   \
    [2138, 8],   \
    [2139, 4],   \
    [2140, 17],   \
    [2141, 8],   \
    [2142, 16],   \
    [2143, 3],   \
    [2144, 13],   \
    [2145, 10],   \
    [2146, 14],   \
    [2147, 5],   \
    [2149, 9],   \
    [2150, 1],   \
    [2151, 17],   \
    [2152, 14],   \
    [2153, 7],   \
    [2154, 12],   \
    [2155, 15],   \
    [2156, 4],   \
    [2158, 8],   \
    [2159, 10],   \
    [2160, 12],   \
    [2161, 7],   \
    [2163, 4],   \
    [2164, 5],   \
    [2165, 5],   \
    [2166, 10],   \
    [2167, 2],   \
    [2168, 15],   \
    [2169, 12],   \
    [2170, 3],   \
    [2172, 17],   \
    [2173, 10],   \
    [2174, 16],   \
    [2175, 16],   \
    [2176, 1],   \
    [2178, 4],   \
    [2179, 8],   \
    [2180, 3],   \
    [2181, 1],   \
    [2182, 16],   \
    [2183, 10],   \
    [2184, 2],   \
    [2185, 16],   \
    [2186, 16],   \
    [2189, 10],   \
    [2190, 16],   \
    [2191, 9],   \
    [2193, 15],   \
    [2195, 16],   \
    [2196, 7],   \
    [2198, 5],   \
    [2199, 5],   \
    [2200, 8],   \
    [2201, 14],   \
    [2202, 7],   \
    [2203, 7],   \
    [2204, 6],   \
    [2205, 3],   \
    [2206, 16],   \
    [2207, 6],   \
    [2208, 13],   \
    [2209, 3],   \
    [2210, 4],   \
    [2212, 6],   \
    [2213, 9],   \
    [2214, 5],   \
    [2215, 5],   \
    [2216, 3],   \
    [2217, 3],   \
    [2218, 13],   \
    [2219, 17],   \
    [2220, 17],   \
    [2221, 7],   \
    [2222, 16],   \
    [2223, 12],   \
    [2224, 11],   \
    [2225, 1],   \
    [2226, 4],   \
    [2227, 14],   \
    [2228, 14],   \
    [2229, 7],   \
    [2230, 12],   \
    [2231, 16],   \
    [2232, 3],   \
    [2233, 4],   \
    [2234, 14],   \
    [2236, 5],   \
    [2237, 6],   \
    [2238, 7],   \
    [2239, 17],   \
    [2240, 13],   \
    [2241, 16],   \
    [2242, 4],   \
    [2243, 13],   \
    [2244, 14],   \
    [2245, 13],   \
    [2246, 14],   \
    [2247, 10],   \
    [2248, 16],   \
    [2249, 4],   \
    [2250, 2],   \
    [2251, 3],   \
    [2252, 15],   \
    [2253, 3],   \
    [2257, 12],   \
    [2258, 11],   \
    [2259, 13],   \
    [2260, 17],   \
    [2261, 5],   \
    [2262, 12],   \
    [2263, 14],   \
    [2264, 14],   \
    [2265, 12],   \
    [2266, 1],   \
    [2267, 8],   \
    [2268, 3],   \
    [2269, 2],   \
    [2270, 9],   \
    [2271, 10],   \
    [2272, 3],   \
    [2273, 7],   \
    [2274, 1],   \
    [2275, 6],   \
    [2276, 5],   \
    [2277, 4],   \
    [2279, 9],   \
    [2282, 3],   \
    [2283, 14],   \
    [2284, 8],   \
    [2286, 8],   \
    [2287, 7],   \
    [2288, 15],   \
    [2289, 15],   \
    [2291, 11],   \
    [2293, 5],   \
    [2294, 1],   \
    [2295, 9],   \
    [2296, 17],   \
    [2297, 4],   \
    [2298, 7],   \
    [2299, 5],   \
    [2300, 10],   \
    [2301, 3],   \
    [2302, 11],   \
    [2303, 7],   \
    [2304, 14],   \
    [2305, 11],   \
    [2306, 1],   \
    [2307, 13],   \
    [2308, 4],   \
    [2309, 5],   \
    [2310, 5],   \
    [2311, 1],   \
    [2312, 10],   \
    [2313, 3],   \
    [2314, 14],   \
    [2315, 16],   \
    [2316, 5],   \
    [2317, 6],   \
    [2318, 14],   \
    [2319, 7],   \
    [2320, 16],   \
    [2321, 17],   \
    [2323, 2],   \
    [2324, 1],   \
    [2326, 6],   \
    [2327, 2],   \
    [2328, 12],   \
    [2329, 12],   \
    [2331, 6],   \
    [2332, 16],   \
    [2333, 7],   \
    [2334, 6],   \
    [2335, 6],   \
    [2336, 11],   \
    [2337, 2],   \
    [2338, 1],   \
    [2339, 16],   \
    [2340, 7],   \
    [2341, 6],   \
    [2342, 14],   \
    [2343, 7],   \
    [2344, 6],   \
    [2345, 14],   \
    [2346, 14],   \
    [2347, 11],   \
    [2348, 8],   \
    [2349, 16],   \
    [2350, 6],   \
    [2352, 8],   \
    [2353, 11],   \
    [2354, 5],   \
    [2355, 2],   \
    [2356, 2],   \
    [2357, 14],   \
    [2358, 14],   \
    [2359, 17],   \
    [2360, 7],   \
    [2361, 9],   \
    [2362, 6],   \
    [2363, 10],   \
    [2364, 15],   \
    [2365, 14],   \
    [2367, 17],   \
    [2368, 5],   \
    [2369, 7],   \
    [2370, 1],   \
    [2371, 3],   \
    [2372, 16],   \
    [2373, 14],   \
    [2375, 3],   \
    [2376, 12],   \
    [2377, 7],   \
    [2378, 5],   \
    [2379, 5],   \
    [2380, 16],   \
    [2381, 5],   \
    [2382, 7],   \
    [2383, 10],   \
    [2385, 9],   \
    [2386, 4],   \
    [2387, 6],   \
    [2388, 17],   \
    [2389, 3],   \
    [2390, 15],   \
    [2391, 13],   \
    [2392, 5],   \
    [2393, 16],   \
    [2394, 12],   \
    [2396, 15],   \
    [2397, 3],   \
    [2398, 12],   \
    [2400, 10],   \
    [2401, 4],   \
    [2402, 17],   \
    [2404, 4],   \
    [2406, 17],   \
    [2407, 11],   \
    [2409, 7],   \
    [2410, 8],   \
    [2411, 3],   \
    [2412, 9],   \
    [2413, 3],   \
    [2414, 17],   \
    [2415, 11],   \
    [2417, 3],   \
    [2418, 13],   \
    [2419, 6],   \
    [2420, 2],   \
    [2422, 5],   \
    [2423, 5],   \
    [2424, 6],   \
    [2425, 9],   \
    [2426, 15],   \
    [2427, 7],   \
    [2428, 2],   \
    [2429, 6],   \
    [2430, 4],   \
    [2431, 3],   \
    [2432, 3],   \
    [2434, 9],   \
    [2437, 6],   \
    [2439, 11],   \
    [2440, 5],   \
    [2441, 12],   \
    [2443, 5],   \
    [2444, 6],   \
    [2445, 7],   \
    [2447, 17],   \
    [2448, 6],   \
    [2449, 17],   \
    [2450, 10],   \
    [2452, 4],   \
    [2453, 8],   \
    [2454, 4],   \
    [2455, 1],   \
    [2456, 14],   \
    [2458, 11],   \
    [2459, 2],   \
    [2460, 3],   \
    [2461, 10],   \
    [2462, 15],   \
    [2464, 2],   \
    [2465, 13],   \
    [2466, 7],   \
    [2467, 15],   \
    [2468, 7],   \
    [2469, 13],   \
    [2470, 10],   \
    [2471, 10],   \
    [2472, 16],   \
    [2473, 17],   \
    [2474, 3],   \
    [2475, 11],   \
    [2478, 14],   \
    [2479, 9],   \
    [2480, 4],   \
    [2481, 11],   \
    [2482, 16],   \
    [2484, 15],   \
    [2485, 10],   \
    [2486, 5],   \
    [2487, 10],   \
    [2488, 3],   \
    [2489, 1],   \
    [2490, 15],   \
    [2491, 9],   \
    [2492, 6],   \
    [2493, 11],   \
    [2494, 5],   \
    [2495, 15],   \
    [2497, 15],   \
    [2498, 12],   \
    [2499, 3],   \
    [2500, 9],   \
    [2502, 13],   \
    [2503, 7],   \
    [2504, 1],   \
    [2505, 6],   \
    [2506, 8],   \
    [2507, 3],   \
    [2508, 3],   \
    [2509, 3],   \
    [2510, 1],   \
    [2511, 8],   \
    [2512, 1],   \
    [2513, 11],   \
    [2514, 11],   \
    [2515, 11],   \
    [2516, 7],   \
    [2517, 3],   \
    [2518, 6],   \
    [2519, 11],   \
    [2520, 8],   \
    [2521, 7],   \
    [2522, 5],   \
    [2523, 16],   \
    [2524, 16],   \
    [2525, 12],   \
    [2526, 1],   \
    [2527, 2],   \
    [2528, 15],   \
    [2529, 8],   \
    [2530, 6],   \
    [2531, 14],   \
    [2532, 7],   \
    [2533, 11],   \
    [2534, 5],   \
    [2535, 14],   \
    [2536, 4],   \
    [2537, 5],   \
    [2538, 2],   \
    [2539, 5],   \
    [2540, 8],   \
    [2541, 13],   \
    [2542, 16],   \
    [2543, 7],   \
    [2544, 16],   \
    [2545, 14],   \
    [2546, 1],   \
    [2547, 10],   \
    [2548, 4],   \
    [2549, 6],   \
    [2550, 4],   \
    [2551, 13],   \
    [2552, 15],   \
    [2553, 1],   \
    [2554, 1],   \
    [2555, 2],   \
    [2556, 9],   \
    [2557, 14],   \
    [2558, 8],   \
    [2559, 14],   \
    [2560, 10],   \
    [2561, 9],   \
    [2562, 13],   \
    [2563, 14],   \
    [2564, 2],   \
    [2565, 14],   \
    [2566, 4],   \
    [2567, 9],   \
    [2568, 10],   \
    [2569, 6],   \
    [2570, 12],   \
    [2571, 7],   \
    [2572, 11],   \
    [2573, 8],   \
    [2574, 8],   \
    [2575, 10],   \
    [2576, 1],   \
    [2577, 3],   \
    [2578, 6],   \
    [2579, 14],   \
    [2580, 14],   \
    [2581, 11],   \
    [2582, 10],   \
    [2584, 2],   \
    [2585, 10],   \
    [2586, 5],   \
    [2587, 16],   \
    [2588, 8],   \
    [2589, 11],   \
    [2595, 17],   \
    [2596, 2],   \
    [2597, 3],   \
    [2598, 8],   \
    [2599, 6],   \
    [2601, 13],   \
    [2602, 14],   \
    [2604, 1],   \
    [2605, 4],   \
    [2611, 13],   \
    [2612, 6],   \
    [2613, 16],   \
    [2617, 14],   \
    [2624, 1],   \
    [2632, 13],   \
    [2634, 11],   \
    [2636, 4],   \
    [2637, 3],   \
    [2638, 7],   \
    [2639, 9],   \
    [2640, 4],   \
    [2641, 14],   \
    [2642, 16],   \
    [2643, 16],   \
    [2644, 4],   \
    [2645, 3],   \
    [2646, 4],   \
    [2647, 16],   \
    [2648, 1],   \
    [2649, 16],   \
    [2650, 7],   \
    [2651, 12],   \
    [2652, 9],   \
    [2653, 15],   \
    [2655, 6],   \
    [2657, 4],   \
    [2658, 16],   \
    [2659, 17],   \
    [2660, 16],   \
    [2661, 9],   \
    [2662, 14],   \
    [2663, 1],   \
    [2664, 11],   \
    [2666, 16],   \
    [2667, 8],   \
    [2668, 5],   \
    [2669, 16],   \
    [2670, 7],   \
    [2671, 9],   \
    [2672, 11],   \
    [2673, 11],   \
    [2674, 9],   \
    [2675, 3],   \
    [2676, 5],   \
    [2677, 11],   \
    [2678, 8],   \
    [2679, 2],   \
    [2680, 12],   \
    [2681, 16],   \
    [2682, 10],   \
    [2683, 16],   \
    [2684, 4],   \
    [2685, 9],   \
    [2686, 15],   \
    [2687, 15],   \
    [2688, 2],   \
    [2690, 2],   \
    [2691, 10],   \
    [2692, 15],   \
    [2693, 15],   \
    [2694, 2],   \
    [2695, 12],   \
    [2696, 13],   \
    [2697, 11],   \
    [2698, 5],   \
    [2699, 16],   \
    [2700, 5],   \
    [2701, 11],   \
    [2702, 12],   \
    [2703, 10],   \
    [2704, 16],   \
    [2705, 11],   \
    [2706, 3],   \
    [2707, 3],   \
    [2708, 10],   \
    [2709, 9],   \
    [2710, 15],   \
    [2711, 11],   \
    [2712, 6],   \
    [2713, 1],   \
    [2714, 13],   \
    [2715, 1],   \
    [2716, 1],   \
    [2717, 4],   \
    [2718, 11],   \
    [2719, 2],   \
    [2720, 15],   \
    [2721, 7],   \
    [2722, 12],   \
    [2723, 9],   \
    [2724, 6],   \
    [2725, 11],   \
    [2726, 9],   \
    [2727, 17],   \
    [2728, 15],   \
    [2729, 16],   \
    [2730, 2],   \
    [2731, 5],   \
    [2732, 7],   \
    [2733, 7],   \
    [2734, 1],   \
    [2735, 8],   \
    [2736, 3],   \
    [2737, 13],   \
    [2738, 12],   \
    [2739, 17],   \
    [2740, 2],   \
    [2741, 3],   \
    [2742, 12],   \
    [2743, 14],   \
    [2744, 9],   \
    [2745, 5],   \
    [2746, 6],   \
    [2747, 11],   \
    [2748, 8],   \
    [2749, 12],   \
    [2750, 9],   \
    [2751, 3],   \
    [2752, 9],   \
    [2753, 1],   \
    [2754, 16],   \
    [2755, 16],   \
    [2756, 5],   \
    [2757, 12],   \
    [2758, 12],   \
    [2759, 16],   \
    [2760, 2],   \
    [2761, 7],   \
    [2762, 8],   \
    [2763, 9],   \
    [2764, 10],   \
    [2765, 11],   \
    [2766, 2],   \
    [2767, 10],   \
    [2768, 13],   \
    [2769, 10],   \
    [2770, 16],   \
    [2771, 1],   \
    [2772, 12],   \
    [2773, 11],   \
    [2774, 5],   \
    [2775, 2],   \
    [2776, 14],   \
    [2777, 16],   \
    [2778, 12],   \
    [2779, 11],   \
    [2780, 5],   \
    [2781, 13],   \
    [2782, 8],   \
    [2783, 6],   \
    [2784, 14],   \
    [2785, 13],   \
    [2786, 4],   \
    [2787, 7],   \
    [2788, 13],   \
    [2789, 1],   \
    [2790, 11],   \
    [2791, 16],   \
    [2792, 1],   \
    [2793, 11],   \
    [2794, 1],   \
    [2795, 6],   \
    [2796, 10],   \
    [2797, 11],   \
    [2798, 4],   \
    [2799, 15],   \
    [2800, 9],   \
    [2801, 9],   \
    [2802, 5],   \
    [2803, 8],   \
    [2804, 14],   \
    [2805, 4],   \
    [2806, 16],   \
    [2807, 10],   \
    [2808, 9],   \
    [2809, 16],   \
    [2810, 9],   \
    [2811, 4],   \
    [2812, 13],   \
    [2813, 11],   \
    [2814, 16],   \
    [2815, 2],   \
    [2816, 12],   \
    [2817, 5],   \
    [2818, 13],   \
    [2819, 11],   \
    [2820, 10],   \
    [2821, 5],   \
    [2822, 12],   \
    [2823, 12],   \
    [2824, 3],   \
    [2825, 4],   \
    [2826, 7],   \
    [2827, 5],   \
    [2828, 10],   \
    [2829, 7],   \
    [2830, 6],   \
    [2831, 7],   \
    [2832, 17],   \
    [2833, 17],   \
    [2834, 3],   \
    [2835, 8],   \
    [2836, 4],   \
    [2837, 11],   \
    [2838, 16],   \
    [2839, 17],   \
    [2840, 15],   \
    [2841, 4],   \
    [2842, 4],   \
    [2843, 14],   \
    [2844, 3],   \
    [2845, 3],   \
    [2846, 7],   \
    [2847, 10],   \
    [2848, 14],   \
    [2849, 5],   \
    [2850, 9],   \
    [2851, 5],   \
    [2852, 8],   \
    [2853, 3],   \
    [2854, 17],   \
    [2855, 15],   \
    [2856, 14],   \
    [2857, 13],   \
    [2858, 12],   \
    [2859, 17],   \
    [2860, 6],   \
    [2861, 11],   \
    [2862, 7],   \
    [2863, 7],   \
    [2864, 14],   \
    [2865, 2],   \
    [2866, 2],   \
    [2867, 4],   \
    [2868, 13],   \
    [2869, 2],   \
    [2870, 5],   \
    [2871, 11],   \
    [2872, 2],   \
    [2873, 4],   \
    [2874, 4],   \
    [2875, 5],   \
    [2876, 8],   \
    [2877, 5],   \
    [2878, 17],   \
    [2879, 10],   \
    [2880, 17],   \
    [2881, 5],   \
    [2882, 7],   \
    [2883, 1],   \
    [2884, 6],   \
    [2885, 12],   \
    [2886, 7],   \
    [2887, 5],   \
    [2890, 7],   \
    [2891, 9],   \
    [2892, 16],   \
    [2893, 11],   \
    [2894, 2],   \
    [2895, 9],   \
    [2896, 16],   \
    [2897, 6],   \
    [2898, 15],   \
    [2899, 4],   \
    [2900, 2],   \
    [2902, 17],   \
    [2903, 11],   \
    [2904, 11],   \
    [2905, 8],   \
    [2906, 16],   \
    [2907, 14],   \
    [2908, 12],   \
    [2909, 17],   \
    [2910, 17],   \
    [2911, 15],   \
    [2912, 10],   \
    [2913, 1],   \
    [2914, 15],   \
    [2915, 12],   \
    [2916, 1],   \
    [2917, 11],   \
    [2918, 5],   \
    [2919, 16],   \
    [2920, 9],   \
    [2921, 13],   \
    [2922, 3],   \
    [2923, 16],   \
    [2924, 1],   \
    [2925, 9],   \
    [2926, 11],   \
    [2927, 2],   \
    [2928, 9],   \
    [2929, 8],   \
    [2930, 8],   \
    [2931, 6],   \
    [2932, 4],   \
    [2933, 3],   \
    [2934, 16],   \
    [2935, 15],   \
    [2936, 1],   \
    [2938, 9],   \
    [2939, 4],   \
    [2940, 12],   \
    [2941, 13],   \
    [2942, 17],   \
    [2943, 17],   \
    [1984, 9],   \
    [1985, 1],   \
    [1986, 17],   \
    [1987, 3],   \
    [1988, 7],   \
    [1989, 3],   \
    [1990, 16],   \
    [1991, 6],   \
    [1992, 7],   \
    [1993, 2],   \
    [1994, 16],   \
    [1995, 16],   \
    [1996, 7],   \
    [1997, 5],   \
    [1998, 5],   \
    [1999, 13],   \
    [2000, 14],   \
    [2001, 1],   \
    [2002, 13],   \
    [2004, 12],   \
    [2005, 10],   \
    [2007, 13],   \
    [2008, 15],   \
    [2009, 15],   \
    [2010, 12],   \
    [2011, 9],   \
    [2012, 9],   \
    [2013, 12],   \
    [2014, 15],   \
    [2015, 1],   \
    [2016, 7],   \
    [2017, 17],   \
    [2018, 17],   \
    [2019, 5],   \
    [2020, 4],   \
    [2021, 4],   \
    [2022, 14],   \
    [2023, 2],   \
    [2024, 4],   \
    [2025, 11],   \
    [2026, 14],   \
    [2027, 10],   \
    [2028, 7],   \
    [2029, 13],   \
    [2030, 5],   \
    [2031, 14],   \
    [2032, 4],   \
    [2033, 3],   \
    [2034, 7],   \
    [2035, 6],   \
    [2036, 13],   \
    [2037, 4],   \
    [2038, 14],   \
    [2039, 7],   \
    [2040, 17],   \
    [2041, 11],   \
    [2042, 13],   \
    [2043, 10],   \
    [2044, 4],   \
    [2045, 12],   \
    [2046, 6],   \
    [2047, 12],   \
    [2048, 2],   \
    [2049, 4],   \
    [2050, 8],   \
    [2051, 10],   \
    [2052, 16],   \
    [2053, 15],   \
    [2054, 8],   \
    [2055, 4],   \
    [2056, 2],   \
    [2057, 2],   \
    [2058, 14],   \
    [2059, 2],   \
    [2060, 1],   \
    [2061, 9],   \
    [2062, 8],   \
    [2063, 1],   \
    [2064, 15],   \
    [2065, 11],   \
    [2068, 16],   \
    [2069, 16],   \
    [2070, 3],   \
    [2071, 4],   \
    [2072, 10],   \
    [2073, 12],   \
    [2074, 9],   \
    [2075, 11],   \
    [2076, 8],   \
    [2077, 12],   \
    [2078, 7],   \
    [2079, 15],   \
    [2080, 5],   \
    [2081, 3],   \
    [2082, 2],   \
    [2083, 13],   \
    [2084, 15],   \
    [2085, 10],   \
    [2086, 7],   \
    [2087, 4],   \
    [2088, 1],   \
    [2089, 7],   \
    [2090, 10],   \
    [2091, 7],   \
    [2092, 17],   \
    [2093, 12],   \
    [2094, 2],   \
    [2095, 1],   \
    [2096, 4],   \
    [2097, 12],   \
    [2098, 4],   \
    [2099, 3],   \
    [2100, 2],   \
    [2101, 13],   \
    [2103, 8],   \
    [2104, 14],   \
    [2105, 15],   \
    [2106, 5],   \
    [2107, 11],   \
    [2108, 4],   \
    [2109, 1],   \
    [2110, 5],   \
    [2111, 6],   \
    [2113, 16],   \
    [2114, 15],   \
    [2115, 15],   \
    [2116, 16],   \
    [2117, 11],   \
    [2118, 8],   \
    [2119, 14],   \
    [2120, 2],   \
    [2121, 2],   \
    [2122, 1],   \
    [2123, 5],   \
    [2124, 5],   \
    [2125, 12],   \
    [2126, 14],   \
    [2127, 10],   \
    [2128, 5],   \
    [2129, 8],   \
    [2130, 16],   \
    [2131, 9],   \
    [2132, 16],   \
    [2133, 3],   \
    [2134, 8],   \
    [2135, 15],   \
    [2136, 1],   \
    [2137, 1],   \
    [2138, 13],   \
    [2140, 13],   \
    [2141, 1],   \
    [2142, 5],   \
    [2143, 1],   \
    [2144, 4],   \
    [2145, 12],   \
    [2146, 5],   \
    [2147, 3],   \
    [2148, 5],   \
    [2149, 13],   \
    [2150, 6],   \
    [2151, 3],   \
    [2152, 9],   \
    [2153, 13],   \
    [2154, 15],   \
    [2155, 6],   \
    [2156, 6],   \
    [2157, 14],   \
    [2158, 12],   \
    [2159, 7],   \
    [2160, 8],   \
    [2161, 17],   \
    [2162, 13],   \
    [2163, 6],   \
    [2164, 7],   \
    [2165, 2],   \
    [2166, 5],   \
    [2167, 8],   \
    [2168, 17],   \
    [2169, 7],   \
    [2170, 14],   \
    [2171, 7],   \
    [2172, 10],   \
    [2174, 13],   \
    [2175, 5],   \
    [2176, 5],   \
    [2177, 12],   \
    [2178, 13],   \
    [2179, 10],   \
    [2180, 12],   \
    [2181, 12],   \
    [2182, 3],   \
    [2183, 13],   \
    [2184, 3],   \
    [2185, 14],   \
    [2186, 3],   \
    [2187, 7],   \
    [2188, 5],   \
    [2189, 2],   \
    [2190, 4],   \
    [2191, 17],   \
    [2192, 13],   \
    [2193, 14],   \
    [2194, 6],   \
    [2195, 13],   \
    [2197, 14],   \
    [2198, 1],   \
    [2200, 14],   \
    [2201, 4],   \
    [2202, 17],   \
    [2203, 14],   \
    [2204, 8],   \
    [2205, 2],   \
    [2206, 1],   \
    [2207, 12],   \
    [2208, 4],   \
    [2209, 13],   \
    [2210, 6],   \
    [2211, 5],   \
    [2212, 11],   \
    [2213, 14],   \
    [2214, 1],   \
    [2215, 15],   \
    [2216, 11],   \
    [2217, 14],   \
    [2218, 2],   \
    [2219, 2],   \
    [2220, 10],   \
    [2221, 10],   \
    [2222, 2],   \
    [2223, 4],   \
    [2224, 15],   \
    [2225, 13],   \
    [2226, 10],   \
    [2227, 8],   \
    [2228, 13],   \
    [2229, 15],   \
    [2230, 3],   \
    [2231, 1],   \
    [2232, 6],   \
    [2233, 12],   \
    [2234, 17],   \
    [2235, 5],   \
    [2236, 3],   \
    [2237, 10],   \
    [2238, 13],   \
    [2239, 13],   \
    [2240, 2],   \
    [2241, 11],   \
    [2242, 13],   \
    [2243, 10],   \
    [2244, 4],   \
    [2245, 10],   \
    [2246, 2],   \
    [2247, 3],   \
    [2248, 1],   \
    [2249, 8],   \
    [2250, 16],   \
    [2251, 4],   \
    [2252, 13],   \
    [2253, 17],   \
    [2257, 7],   \
    [2258, 12],   \
    [2259, 1],   \
    [2260, 2],   \
    [2261, 1],   \
    [2262, 11],   \
    [2263, 7],   \
    [2264, 8],   \
    [2265, 7],   \
    [2266, 16],   \
    [2267, 4],   \
    [2268, 9],   \
    [2269, 13],   \
    [2270, 4],   \
    [2271, 8],   \
    [2272, 9],   \
    [2273, 13],   \
    [2274, 6],   \
    [2275, 9],   \
    [2276, 2],   \
    [2277, 11],   \
    [2278, 6],   \
    [2279, 14],   \
    [2281, 16],   \
    [2282, 11],   \
    [2283, 1],   \
    [2284, 4],   \
    [2285, 16],   \
    [2286, 12],   \
    [2287, 9],   \
    [2288, 3],   \
    [2289, 17],   \
    [2290, 17],   \
    [2291, 13],   \
    [2292, 2],   \
    [2293, 7],   \
    [2294, 14],   \
    [2295, 17],   \
    [2296, 11],   \
    [2297, 6],   \
    [2298, 12],   \
    [2299, 2],   \
    [2300, 9],   \
    [2301, 13],   \
    [2302, 8],   \
    [2303, 1],   \
    [2304, 7],   \
    [2305, 16],   \
    [2306, 5],   \
    [2307, 6],   \
    [2308, 2],   \
    [2309, 11],   \
    [2310, 12],   \
    [2311, 14],   \
    [2312, 2],   \
    [2313, 15],   \
    [2314, 4],   \
    [2315, 11],   \
    [2316, 15],   \
    [2317, 16],   \
    [2318, 8],   \
    [2319, 3],   \
    [2320, 9],   \
    [2321, 5],   \
    [2322, 4],   \
    [2323, 4],   \
    [2324, 3],   \
    [2325, 3],   \
    [2326, 3],   \
    [2327, 7],   \
    [2328, 2],   \
    [2329, 1],   \
    [2330, 7],   \
    [2331, 17],   \
    [2332, 17],   \
    [2333, 8],   \
    [2334, 3],   \
    [2335, 5],   \
    [2336, 17],   \
    [2337, 16],   \
    [2338, 9],   \
    [2339, 17],   \
    [2340, 5],   \
    [2341, 12],   \
    [2342, 11],   \
    [2343, 5],   \
    [2344, 9],   \
    [2345, 13],   \
    [2346, 2],   \
    [2347, 10],   \
    [2348, 7],   \
    [2349, 9],   \
    [2350, 13],   \
    [2352, 12],   \
    [2353, 9],   \
    [2354, 2],   \
    [2355, 13],   \
    [2356, 5],   \
    [2357, 7],   \
    [2358, 4],   \
    [2359, 3],   \
    [2360, 17],   \
    [2361, 17],   \
    [2362, 7],   \
    [2363, 6],   \
    [2364, 5],   \
    [2365, 12],   \
    [2366, 14],   \
    [2367, 5],   \
    [2368, 4],   \
    [2369, 6],   \
    [2370, 8],   \
    [2371, 8],   \
    [2373, 13],   \
    [2376, 8],   \
    [2377, 6],   \
    [2378, 11],   \
    [2379, 8],   \
    [2380, 2],   \
    [2381, 4],   \
    [2382, 6],   \
    [2383, 14],   \
    [2384, 6],   \
    [2385, 4],   \
    [2386, 13],   \
    [2387, 1],   \
    [2388, 16],   \
    [2389, 6],   \
    [2390, 11],   \
    [2391, 9],   \
    [2392, 13],   \
    [2393, 17],   \
    [2394, 17],   \
    [2395, 15],   \
    [2396, 10],   \
    [2397, 1],   \
    [2398, 11],   \
    [2399, 10],   \
    [2400, 6],   \
    [2401, 9],   \
    [2402, 12],   \
    [2403, 5],   \
    [2404, 3],   \
    [2405, 15],   \
    [2406, 4],   \
    [2407, 7],   \
    [2408, 12],   \
    [2409, 8],   \
    [2410, 3],   \
    [2411, 1],   \
    [2412, 2],   \
    [2413, 11],   \
    [2414, 14],   \
    [2415, 1],   \
    [2416, 11],   \
    [2417, 17],   \
    [2418, 6],   \
    [2419, 10],   \
    [2420, 8],   \
    [2421, 6],   \
    [2422, 7],   \
    [2423, 7],   \
    [2424, 8],   \
    [2426, 7],   \
    [2427, 4],   \
    [2428, 12],   \
    [2429, 7],   \
    [2431, 16],   \
    [2432, 5],   \
    [2433, 17],   \
    [2434, 3],   \
    [2435, 4],   \
    [2436, 7],   \
    [2437, 15],   \
    [2438, 7],   \
    [2440, 2],   \
    [2441, 17],   \
    [2442, 17],   \
    [2443, 1],   \
    [2444, 14],   \
    [2445, 4],   \
    [2446, 4],   \
    [2447, 9],   \
    [2448, 14],   \
    [2449, 11],   \
    [2450, 16],   \
    [2451, 9],   \
    [2452, 12],   \
    [2453, 13],   \
    [2454, 2],   \
    [2455, 16],   \
    [2456, 13],   \
    [2457, 1],   \
    [2458, 12],   \
    [2459, 14],   \
    [2460, 12],   \
    [2461, 7],   \
    [2462, 14],   \
    [2464, 14],   \
    [2465, 6],   \
    [2466, 15],   \
    [2467, 9],   \
    [2468, 16],   \
    [2469, 6],   \
    [2470, 17],   \
    [2471, 15],   \
    [2472, 7],   \
    [2473, 12],   \
    [2474, 9],   \
    [2475, 16],   \
    [2478, 16],   \
    [2479, 11],   \
    [2480, 6],   \
    [2481, 10],   \
    [2482, 9],   \
    [2484, 12],   \
    [2485, 9],   \
    [2486, 15],   \
    [2487, 2],   \
    [2488, 11],   \
    [2489, 2],   \
    [2490, 5],   \
    [2491, 6],   \
    [2492, 16],   \
    [2493, 10],   \
    [2494, 13],   \
    [2495, 11],   \
    [2496, 9],   \
    [2497, 14],   \
    [2498, 6],   \
    [2499, 15],   \
    [2500, 5],   \
    [2502, 8],   \
    [2503, 2],   \
    [2504, 5],   \
    [2505, 9],   \
    [2506, 6],   \
    [2507, 13],   \
    [2508, 5],   \
    [2509, 6],   \
    [2510, 3],   \
    [2511, 14],   \
    [2512, 9],   \
    [2513, 3],   \
    [2514, 5],   \
    [2515, 7],   \
    [2048, 15],   \
    [2049, 12],   \
    [2050, 6],   \
    [2051, 2],   \
    [2052, 12],   \
    [2053, 13],   \
    [2054, 3],   \
    [2055, 15],   \
    [2056, 8],   \
    [2057, 15],   \
    [2058, 8],   \
    [2059, 8],   \
    [2060, 13],   \
    [2061, 3],   \
    [2062, 10],   \
    [2063, 4],   \
    [2064, 9],   \
    [2065, 17],   \
    [2066, 7],   \
    [2067, 6],   \
    [2068, 14],   \
    [2069, 15],   \
    [2070, 15],   \
    [2071, 15],   \
    [2072, 5],   \
    [2073, 16],   \
    [2074, 17],   \
    [2075, 8],   \
    [2076, 9],   \
    [2077, 9],   \
    [2078, 16],   \
    [2079, 17],   \
    [2080, 3],   \
    [2081, 16],   \
    [2082, 16],   \
    [2083, 12],   \
    [2084, 4],   \
    [2085, 4],   \
    [2086, 13],   \
    [2087, 7],   \
    [2088, 16],   \
    [2089, 5],   \
    [2090, 9],   \
    [2091, 2],   \
    [2092, 16],   \
    [2093, 8],   \
    [2094, 16],   \
    [2095, 13],   \
    [2096, 10],   \
    [2097, 5],   \
    [2098, 13],   \
    [2099, 14],   \
    [2100, 11],   \
    [2101, 16],   \
    [2102, 10],   \
    [2103, 2],   \
    [2104, 10],   \
    [2105, 2],   \
    [2106, 8],   \
    [2107, 5],   \
    [2108, 10],   \
    [2109, 2],   \
    [2110, 14],   \
    [2111, 17],   \
    [2112, 11],   \
    [2113, 6],   \
    [2114, 2],   \
    [2115, 1],   \
    [2116, 4],   \
    [2117, 1],   \
    [2118, 16],   \
    [2119, 12],   \
    [2120, 6],   \
    [2121, 6],   \
    [2122, 9],   \
    [2123, 9],   \
    [2124, 10],   \
    [2125, 4],   \
    [2126, 3],   \
    [2127, 16],   \
    [2128, 14],   \
    [2129, 4],   \
    [2130, 9],   \
    [2131, 13],   \
    [2132, 14],   \
    [2133, 11],   \
    [2134, 15],   \
    [2135, 7],   \
    [2136, 11],   \
    [2137, 3],   \
    [2138, 9],   \
    [2139, 16],   \
    [2140, 3],   \
    [2141, 14],   \
    [2142, 15],   \
    [2143, 15],   \
    [2144, 1],   \
    [2145, 1],   \
    [2146, 4],   \
    [2147, 10],   \
    [2148, 14],   \
    [2149, 10],   \
    [2150, 11],   \
    [2151, 12],   \
    [2152, 4],   \
    [2153, 11],   \
    [2154, 11],   \
    [2155, 1],   \
    [2156, 5],   \
    [2157, 2],   \
    [2158, 1],   \
    [2159, 2],   \
    [2160, 14],   \
    [2161, 11],   \
    [2162, 15],   \
    [2163, 5],   \
    [2164, 9],   \
    [2165, 4],   \
    [2166, 16],   \
    [2167, 14],   \
    [2168, 16],   \
    [2169, 16],   \
    [2170, 6],   \
    [2171, 2],   \
    [2172, 4],   \
    [2173, 4],   \
    [2174, 7],   \
    [2175, 14],   \
    [2176, 9],   \
    [2177, 15],   \
    [2178, 8],   \
    [2179, 4],   \
    [2180, 11],   \
    [2181, 7],   \
    [2182, 7],   \
    [2183, 6],   \
    [2184, 11],   \
    [2185, 6],   \
    [2186, 15],   \
    [2187, 3],   \
    [2188, 1],   \
    [2189, 3],   \
    [2190, 15],   \
    [2191, 4],   \
    [2192, 4],   \
    [2193, 5],   \
    [2194, 16],   \
    [2195, 2],   \
    [2196, 13],   \
    [2197, 15],   \
    [2198, 15],   \
    [2199, 7],   \
    [2200, 12],   \
    [2201, 10],   \
    [2202, 12],   \
    [2203, 17],   \
    [2204, 16],   \
    [2205, 5],   \
    [2206, 17],   \
    [2207, 10],   \
    [2208, 14],   \
    [2209, 8],   \
    [2277, 12],   \
    [2278, 10],   \
    [2279, 2],   \
    [2280, 4],   \
    [2281, 10],   \
    [2282, 4],   \
    [2283, 10],   \
    [2284, 13],   \
    [2285, 12],   \
    [2286, 5],   \
    [2287, 17],   \
    [2288, 5],   \
    [2289, 9],   \
    [2290, 14],   \
    [2291, 15],   \
    [2292, 14],   \
    [2293, 15],   \
    [2294, 4],   \
    [2295, 1],   \
    [2296, 8],   \
    [2297, 1],   \
    [2365, 10],   \
    [2366, 9],   \
    [2367, 7],   \
    [2368, 17],   \
    [2369, 5],   \
    [2370, 7],   \
    [2371, 13],   \
    [2372, 11],   \
    [2373, 8],   \
    [2375, 7],   \
    [2376, 3],   \
    [2377, 5],   \
    [2378, 7],   \
    [2379, 13],   \
    [2380, 13],   \
    [2381, 2],   \
    [2382, 10],   \
    [2383, 3],   \
    [2384, 13],   \
    [2385, 17],   \
    [2386, 2],   \
    [2387, 4],   \
    [2388, 11],   \
    [2389, 14],   \
    [2390, 1],   \
    [2391, 3],   \
    [2392, 4],   \
    [2393, 6],   \
    [2394, 14],   \
    [2395, 4],   \
    [2396, 1],   \
    [2397, 17],   \
    [2398, 10],   \
    [2399, 14],   \
    [2400, 1],   \
    [2401, 16],   \
    [2402, 2],   \
    [2403, 7],   \
    [2404, 10],   \
    [2405, 16],   \
    [2406, 7],   \
    [2407, 16],   \
    [2408, 2],   \
    [2409, 11],   \
    [2410, 13],   \
    [2411, 6],   \
    [2412, 5],   \
    [2413, 14],   \
    [2414, 13],   \
    [2415, 13],   \
    [2416, 4],   \
    [2417, 16],   \
    [2418, 14],   \
    [2419, 2],   \
    [2420, 17],   \
    [2421, 11],   \
    [2422, 4],   \
    [2423, 8],   \
    [2424, 16],   \
    [2425, 2],   \
    [2426, 4],   \
    [2427, 14],   \
    [2428, 13],   \
    [2429, 14],   \
    [2430, 1],   \
    [2431, 9],   \
    [2432, 6],   \
    [2433, 1],   \
    [2434, 2],   \
    [2435, 2],   \
    [2436, 12],   \
    [2437, 9],   \
    [2438, 8],   \
    [2439, 9]]

    cursor.executemany("INSERT INTO teilnehmer VALUES (?,?)",datalist)
    connection.commit()
    connection.close()
    
def andere_tabellen_fuellen():
    connection = sq.connect("sportverein.db")
    cursor = connection.cursor()

    # ------- Daten für Tabelle bank
    datalist = [   \
    [18750023, 'Volksbank Mawicke'],\
    [37500046, 'Sparkasse Echthausen'],\
    [56250069, 'Volksbank Unna'],\
    [75000092, 'Sparkasse Rhynern'],\
    [93750115, 'Deutsche Bank Soest']]

    cursor.executemany("INSERT INTO bank VALUES (?,?)",datalist)
    connection.commit()

    # ------- Daten für Tabelle uebungsgruppe
    datalist = [[1, 'Schwimmen', 1990],   \
    [2, 'Fußball', 2010],   \
    [3, 'Tennis', 2229],   \
    [4, 'Volleyball', 2832],   \
    [5, 'Wasserball', 1990],   \
    [6, 'Tauchen', 2595],   \
    [7, 'Basketball', 2577],   \
    [8, 'Turnen', 2643],   \
    [9, 'Langlauf', 2349],   \
    [10, 'Fußball', 1977],   \
    [11, 'Volleyball', 1993],   \
    [12, 'Schwimmen', 2173],   \
    [13, 'Fußball', 2010],   \
    [14, 'Volleyball', 1993],   \
    [15, 'Fußball', 2189],   \
    [16, 'Schwimmen', 2638],   \
    [17, 'Fußball', 2632]] 

    cursor.executemany("INSERT INTO uebungsgruppe VALUES (?,?,?)",datalist)
    connection.commit()

    # ------- Daten für Tabelle trainingsort
    datalist = [['Bad1', 1],   \
    ['Bad2', 2],   \
    ['Bad3', 2],   \
    ['Halle1', 2],   \
    ['Halle2', 2],   \
    ['Halle3', 3],   \
    ['Platz1', 3],   \
    ['Platz2', 1]]

    cursor.executemany("INSERT INTO trainingsort VALUES (?,?)",datalist)
    connection.commit()

    # ------- Daten für Tabelle trainingseinheit
    datalist = [['Montag', '18:30:00', 1, 'Bad1'],   \
    ['Mittwoch', '19:00:00', 2, 'Platz1'],   \
    ['Dienstag', '17:00:00', 3, 'Platz2'],   \
    ['Donnerstag', '18:30:00', 4, 'Halle2'],   \
    ['Donnerstag', '17:30:00', 5, 'Bad2'],   \
    ['Dienstag', '20:00:00', 6, 'Bad3'],   \
    ['Dienstag', '19:00:00', 7, 'Halle1'],   \
    ['Montag', '18:00:00', 8, 'Halle3'],   \
    ['Mittwoch', '18:00:00', 9, 'Platz1'],   \
    ['Dienstag', '18:30:00', 10, 'Platz1'],   \
    ['Mittwoch', '19:00:00', 11, 'Halle2'],   \
    ['Mittwoch', '19:00:00', 12, 'Bad3'],   \
    ['Donnerstag', '17:30:00', 13, 'Platz1'],   \
    ['Mittwoch', '21:00:00', 14, 'Halle1'],   \
    ['Donnerstag', '17:30:00', 15, 'Platz1'],   \
    ['Montag', '20:00:00', 16, 'Bad2'],   \
    ['Donnerstag', '18:30:00', 17, 'Platz1'],   \
    ['Montag', '17:00:00', 3, 'Platz2'],   \
    ['Mittwoch', '18:30:00', 4, 'Halle2'],   \
    ['Dienstag', '17:30:00', 5, 'Bad2'],   \
    ['Donnerstag', '20:00:00', 6, 'Bad3'],   \
    ['Donnerstag', '19:00:00', 7, 'Halle2'],   \
    ['Dienstag', '18:00:00', 8, 'Halle3'],   \
    ['Montag', '18:00:00', 9, 'Platz1']]


    cursor.executemany("INSERT INTO trainingseinheit VALUES (?,?,?,?)",datalist)
    connection.commit()

    # ------- Daten für Tabelle hausmeister
    datalist=[
    [1, 'Krause', '02922-555421'],  \
    [2, 'Koschulte', '02922-45512'],  \
    [3, 'Kacmarek', '02928-839630']]

    cursor.executemany("INSERT INTO hausmeister VALUES (?,?,?)",datalist)
    connection.commit()

    connection.close()

def createDB():
    tabellen_anlegen()
    mitglieder_fuellen()
    teilnehmer_fuellen()
    andere_tabellen_fuellen()

createDB()
```
